## Step 2: Inspect the Most Useful Venue Fields

```python
relevant_columns = [
    "NAME",
    "POPOS_ADDRESS",
    "TYPE",
    "Amenities",
    "Seating",
    "Seating_and_Tables",
    "RESTROOMS",
    "Accessibility",
    "FOOD_SERVICE",
    "Indoor",
    "Neighborhoods",
    "the_geom"
]

venue_data[relevant_columns].head()
```

## Check Missing Values in the Main Venue Fields

Before starting the cleaning pipeline, I check how much missing data exists in the most important columns.

This helps me identify which fields may need filling, flagging, or formatting during the ETL process.

```python
venue_data[relevant_columns].isnull().sum().sort_values(ascending=False)
```

## Step 3: Begin the Cleaning Pipeline

Now that I have identified the main missing-value issues, the next step is to start cleaning the venue dataset.

In this step, I create a working copy of the data and standardize the main venue fields so they are easier to use in the later ETL stages.

```python
clean_venues = venue_data[relevant_columns].copy()

clean_venues["NAME"] = clean_venues["NAME"].astype(str).str.strip().str.title()
clean_venues["POPOS_ADDRESS"] = clean_venues["POPOS_ADDRESS"].astype(str).str.strip()
clean_venues["TYPE"] = clean_venues["TYPE"].astype(str).str.strip().str.title()
clean_venues["Neighborhoods"] = clean_venues["Neighborhoods"].astype(str).str.strip().str.title()

clean_venues.head()
```

The initial cleaning step worked correctly. The main venue identification fields now look more standardized and easier to read, especially the venue name, address, type, and neighborhood columns. The preview also confirms that the dataset still contains missing values in several amenity-related fields, which means the next ETL step should focus on handling those missing entries consistently before feature engineering begins.

## Step 4: Handle the Main Missing Venue Features

Now that the core venue fields have been standardized, the next step is to handle the missing values in the amenity-related columns.

For this ETL workflow, I replace missing values in the main yes/no venue feature columns with `"Unknown"` so the pipeline keeps the information explicit instead of leaving null values behind.

```python
fill_unknown_cols = [
    "Amenities",
    "Seating",
    "Seating_and_Tables",
    "RESTROOMS",
    "Accessibility",
    "FOOD_SERVICE",
    "Indoor"
]

for col in fill_unknown_cols:
    clean_venues[col] = clean_venues[col].fillna("Unknown")

clean_venues[fill_unknown_cols].isnull().sum()
```

The missing-value handling step worked successfully. All of the main amenity-related columns now show zero missing values, which means the ETL pipeline has corrected one of the biggest data-quality issues in the dataset. This makes the venue features much easier to use in the later suitability scoring and segmentation steps.

## Step 5: Create a Basic Suitability Score

Now that the core venue fields are cleaned, the next step is to create a simple Suitability Score for lead-generation analysis.

This score gives each venue a starting value based on whether it includes useful corporate-event features such as seating, tables, restrooms, food service, indoor space, and accessibility.

```python
clean_venues["suitability_score"] = (
    (clean_venues["Seating"].astype(str).str.upper() == "Y").astype(int)
    + (clean_venues["Seating_and_Tables"].astype(str).str.upper() == "Y").astype(int)
    + (clean_venues["RESTROOMS"].astype(str).str.upper() == "YES").astype(int)
    + (clean_venues["FOOD_SERVICE"].astype(str).str.upper() == "YES").astype(int)
    + (clean_venues["Indoor"].astype(str).str.upper() == "TRUE").astype(int)
    + (clean_venues["Accessibility"].astype(str).str.contains("Direct", case=False, na=False)).astype(int)
)

clean_venues[[
    "NAME",
    "TYPE",
    "Seating",
    "Seating_and_Tables",
    "RESTROOMS",
    "FOOD_SERVICE",
    "Indoor",
    "Accessibility",
    "suitability_score"
]].head()
```

The Suitability Score was created successfully and is already helping distinguish stronger venue candidates from weaker ones. Venues with more useful corporate-event features, such as seating, tables, restrooms, food service, indoor space, and direct accessibility, receive higher scores. The preview shows that the score creates a practical ranking system that can be used later for venue segmentation and lead prioritization

## Step 6: Standardize the Venue Type into Marketing Segments

Now that the Suitability Score is available, the next step is to group the raw venue types into broader marketing categories.

For this project, I standardize the space types into four segments:

- **Indoor**
- **Outdoor**
- **High-Capacity**
- **Boutique**

This makes the dataset easier to use for corporate event targeting and outreach planning.

```python
def classify_space_type(row):
    type_text = str(row["TYPE"]).lower()
    indoor_text = str(row["Indoor"]).lower()

    if "indoor" in type_text or indoor_text == "true":
        return "Indoor"
    elif "plaza" in type_text or "garden" in type_text or "arcade" in type_text:
        return "Outdoor"
    elif row["suitability_score"] >= 5:
        return "High-Capacity"
    else:
        return "Boutique"

clean_venues["space_type_segment"] = clean_venues.apply(classify_space_type, axis=1)

clean_venues[["NAME", "TYPE", "Indoor", "suitability_score", "space_type_segment"]].head()
```

The space-type segmentation step worked correctly. The raw venue types have now been translated into broader marketing categories that are easier to use for outreach analysis. In this preview, plaza and urban garden spaces are being grouped into the Outdoor segment, while the venue that clearly includes indoor characteristics is being assigned to the Indoor segment. This gives the dataset a more usable structure for corporate venue targeting.

## Step 12: Wrap the ETL Workflow into Reusable Functions

To make the notebook more modular and closer to a reusable ETL pipeline, I now organize the main logic into functions.

This makes it easier to rerun the workflow later if the source file is updated.

```python
def standardize_text_columns(df):
    df = df.copy()
    df["NAME"] = df["NAME"].astype(str).str.strip().str.title()
    df["POPOS_ADDRESS"] = df["POPOS_ADDRESS"].astype(str).str.strip()
    df["TYPE"] = df["TYPE"].astype(str).str.strip().str.title()
    df["Neighborhoods"] = df["Neighborhoods"].astype(str).str.strip().str.title()
    return df


def fill_main_missing_values(df):
    df = df.copy()
    fill_unknown_cols = [
        "Amenities",
        "Seating",
        "Seating_and_Tables",
        "RESTROOMS",
        "Accessibility",
        "FOOD_SERVICE",
        "Indoor"
    ]
    for col in fill_unknown_cols:
        df[col] = df[col].fillna("Unknown")
    return df


def create_suitability_score(df):
    df = df.copy()
    df["suitability_score"] = (
        (df["Seating"].astype(str).str.upper() == "Y").astype(int)
        + (df["Seating_and_Tables"].astype(str).str.upper() == "Y").astype(int)
        + (df["RESTROOMS"].astype(str).str.upper() == "YES").astype(int)
        + (df["FOOD_SERVICE"].astype(str).str.upper() == "YES").astype(int)
        + (df["Indoor"].astype(str).str.upper() == "TRUE").astype(int)
        + (df["Accessibility"].astype(str).str.contains("Direct", case=False, na=False)).astype(int)
    )
    return df


def classify_space_type(row):
    type_text = str(row["TYPE"]).lower()
    indoor_text = str(row["Indoor"]).lower()

    if "indoor" in type_text or indoor_text == "true":
        return "Indoor"
    elif "plaza" in type_text or "garden" in type_text or "arcade" in type_text:
        return "Outdoor"
    elif row["suitability_score"] >= 5:
        return "High-Capacity"
    else:
        return "Boutique"


def add_space_type_segment(df):
    df = df.copy()
    df["space_type_segment"] = df.apply(classify_space_type, axis=1)
    return df
```

## Run the Reusable ETL Functions

Here, I rerun the main cleaning and feature engineering steps through the function-based pipeline.

```python
pipeline_venues = venue_data[relevant_columns].copy()

pipeline_venues = standardize_text_columns(pipeline_venues)
pipeline_venues = fill_main_missing_values(pipeline_venues)
pipeline_venues = create_suitability_score(pipeline_venues)
pipeline_venues = add_space_type_segment(pipeline_venues)

pipeline_venues.head()
```

## Step 13: Standardize Coordinate Information

To strengthen the ETL quality checks, I also extract longitude and latitude from the raw geometry field.

This addresses another data-quality issue by making the location information easier to use in later mapping or CRM workflows.

```python
pipeline_venues["longitude"] = pipeline_venues["the_geom"].astype(str).str.extract(r"\[\s*([-0-9.]+)")
pipeline_venues["latitude"] = pipeline_venues["the_geom"].astype(str).str.extract(r",\s*([-0-9.]+)\s*\]")

pipeline_venues["longitude"] = pd.to_numeric(pipeline_venues["longitude"], errors="coerce")
pipeline_venues["latitude"] = pd.to_numeric(pipeline_venues["latitude"], errors="coerce")

pipeline_venues[["NAME", "the_geom", "longitude", "latitude"]].head()
```

## Check the Parsed Coordinates

Now I confirm whether the coordinate parsing worked correctly.

```python
pipeline_venues[["longitude", "latitude"]].isnull().sum()
```

## Step 14: Create an Estimated Price Tier

The source dataset does not include direct pricing, so I create a simple pricing proxy to support the project’s pricing angle.

This estimated price tier is based on venue quality signals such as suitability score, indoor status, and food service availability. It is not a real market price, but it gives the marketing team a practical way to separate lower-touch venues from more premium venue leads.

```python
def estimate_price_tier(row):
    if row["suitability_score"] >= 5 and str(row["Indoor"]).lower() == "true":
        return "Premium"
    elif row["suitability_score"] >= 4:
        return "Standard"
    else:
        return "Budget"

pipeline_venues["estimated_price_tier"] = pipeline_venues.apply(estimate_price_tier, axis=1)

pipeline_venues[[
    "NAME",
    "suitability_score",
    "Indoor",
    "FOOD_SERVICE",
    "estimated_price_tier"
]].head(10)
```

## Step 15: Update the Final CRM-Ready Venue Table

Now I rebuild the final output table so it also includes the parsed coordinates and the estimated price tier.

```python
final_venues = pipeline_venues[[
    "NAME",
    "POPOS_ADDRESS",
    "TYPE",
    "Neighborhoods",
    "suitability_score",
    "space_type_segment",
    "estimated_price_tier",
    "Amenities",
    "Seating",
    "Seating_and_Tables",
    "RESTROOMS",
    "FOOD_SERVICE",
    "Indoor",
    "Accessibility",
    "longitude",
    "latitude"
]].copy()

final_venues = final_venues.sort_values(
    by=["suitability_score", "Neighborhoods"],
    ascending=[False, True]
)

final_venues.head(10)
```

## Step 16: Create a Polished Neighborhood Density Chart

To improve the presentation quality, I now rebuild the neighborhood chart with sorted bars and value labels.

```python
high_suitability = final_venues[final_venues["suitability_score"] >= 4].copy()

neighborhood_density = (
    high_suitability["Neighborhoods"]
    .value_counts()
    .reset_index()
)
neighborhood_density.columns = ["Neighborhood", "High_Suitability_Venue_Count"]
neighborhood_density = neighborhood_density.sort_values(
    by="High_Suitability_Venue_Count",
    ascending=False
)

plt.figure(figsize=(8, 5))
ax = sns.barplot(
    data=neighborhood_density,
    x="Neighborhood",
    y="High_Suitability_Venue_Count",
    color="teal"
)

for container in ax.containers:
    ax.bar_label(container)

plt.title("High-Suitability Venue Count by Neighborhood", fontsize=14, weight="bold")
plt.xlabel("Neighborhood", fontsize=12)
plt.ylabel("High-Suitability Venue Count", fontsize=12)
plt.show()
```
