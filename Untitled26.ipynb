# Step 2: Clean the skincare dataset

In this step, we keep only the skincare columns needed for the project, remove rows with missing values in key fields, convert important columns to numeric format, and standardize the main text fields.

## Keep only the columns needed for the project

```python
skincare_columns = [
    "Product_Name",
    "Brand",
    "Category",
    "Price_USD",
    "Rating",
    "Number_of_Reviews",
    "Skin_Type",
    "Main_Ingredient",
    "Cruelty_Free",
    "Country_of_Origin"
]

skincare = skincare[skincare_columns].copy()
skincare.head()
```

## Check the dataset shape and missing values

```python
print(skincare.shape)
print(skincare.isnull().sum())
```

## Remove rows with missing values in the main product fields

```python
skincare = skincare.dropna(subset=[
    "Product_Name",
    "Category",
    "Price_USD",
    "Rating",
    "Number_of_Reviews",
    "Skin_Type",
    "Main_Ingredient"
]).copy()

print(skincare.shape)
```

## Remove rows that became invalid after numeric conversion

```python
skincare = skincare.dropna(subset=["Price_USD", "Rating", "Number_of_Reviews"]).copy()
print(skincare.shape)
```

## Standardize the main text fields

```python
skincare["Skin_Type"] = skincare["Skin_Type"].astype(str).str.strip()
skincare["Main_Ingredient"] = skincare["Main_Ingredient"].astype(str).str.strip()
skincare["Category"] = skincare["Category"].astype(str).str.strip()
skincare["Product_Name"] = skincare["Product_Name"].astype(str).str.strip()
```

# Step 4: Create ingredient logic and the Sensitivity Index

In this step, we create product logic using the actual ingredient values that exist in the skincare dataset. This will help us compare products against humidity and UV conditions later in Tableau.

## Check the ingredient values in the dataset

This confirms which ingredient names are actually present before grouping them.

```python
skincare["Main_Ingredient"].drop_duplicates().tolist()
```

## Create Ingredient Group

This groups the real ingredient values into broader treatment families that are easier to analyze in Tableau.

```python
def map_ingredient_group(ingredient):
    ing = str(ingredient).lower().strip()

    if ing == "retinol":
        return "Retinoids"
    elif ing == "vitamin c":
        return "Vitamin C"
    elif ing == "salicylic acid":
        return "BHAs"
    elif ing in ["hyaluronic acid", "glycerin", "aloe vera", "shea butter"]:
        return "Hydrators & Barrier Support"
    else:
        return "Other Actives"

skincare["Ingredient_Group"] = skincare["Main_Ingredient"].apply(map_ingredient_group)
skincare["Ingredient_Group"].value_counts()
```

## Create Photosensitive Flag

This identifies products that may become more reactive under stronger UV conditions.

```python
photosensitive_groups = ["Retinoids", "Vitamin C", "BHAs"]

skincare["Photosensitive_Flag"] = skincare["Ingredient_Group"].isin(photosensitive_groups).astype(int)

skincare[["Main_Ingredient", "Ingredient_Group", "Photosensitive_Flag"]].drop_duplicates()
```

## Create Moisture Loss Risk

This approximates how likely an ingredient group is to struggle under dry or dehydrating conditions.

```python
def moisture_loss_risk(group):
    if group in ["Retinoids", "BHAs"]:
        return 3
    elif group == "Vitamin C":
        return 2
    elif group == "Hydrators & Barrier Support":
        return 1
    else:
        return 2

skincare["Moisture_Loss_Risk"] = skincare["Ingredient_Group"].apply(moisture_loss_risk)

skincare[["Ingredient_Group", "Moisture_Loss_Risk"]].drop_duplicates()
```

## Create a base product efficacy score

This combines product rating and review volume into one simple score that reflects product strength and customer confidence.

```python
skincare["Review_Score"] = np.log1p(skincare["Number_of_Reviews"])

skincare["Base_Efficacy_Score"] = (
    skincare["Rating"] * 0.7 +
    skincare["Review_Score"] * 0.3
)

skincare["Base_Efficacy_Score"] = skincare["Base_Efficacy_Score"].round(2)

skincare[["Rating", "Number_of_Reviews", "Base_Efficacy_Score"]].head()
```

## Create the Sensitivity Index

This score estimates how sensitive a product may be under strong UV and moisture-loss conditions.

```python
skincare["Sensitivity_Index"] = (
    skincare["Photosensitive_Flag"] * 0.5 +
    skincare["Moisture_Loss_Risk"] * 0.5
)

skincare["Sensitivity_Index"] = skincare["Sensitivity_Index"].round(2)

skincare[[
    "Product_Name",
    "Main_Ingredient",
    "Ingredient_Group",
    "Photosensitive_Flag",
    "Moisture_Loss_Risk",
    "Sensitivity_Index"
]].head()
```

# Step 5: Merge the skincare and weather logic into one Tableau-ready dataset

In this step, we combine the cleaned skincare data with the three-region weather data. Because skincare products are not naturally tied to one city, this merge creates a scenario-based comparison table so each product can be evaluated under each climate condition.

## Keep only the final weather fields needed for merging

```python
weather_final = weather_regions[[
    "location_name",
    "country",
    "last_updated",
    "temperature_celsius",
    "humidity",
    "uv_index",
    "condition_text",
    "Region"
]].copy()

weather_final.head()
```

## Create a simple weather mode field

This will help later when we build the Tableau toggle between current weather and seasonal-style comparison logic.

```python
weather_final["Weather_Mode"] = "Current Weather"
weather_final.head()
```

## Create a month field from the weather timestamp

This helps with seasonal views and easier grouping later.

```python
weather_final["Month"] = weather_final["last_updated"].dt.month_name()
weather_final.head()
```

## Create a season field from the month

This gives us a more readable climate grouping for Tableau.

```python
def map_season(month):
    if month in ["December", "January", "February"]:
        return "Winter"
    elif month in ["March", "April", "May"]:
        return "Spring"
    elif month in ["June", "July", "August"]:
        return "Summer"
    else:
        return "Autumn"

weather_final["Season"] = weather_final["Month"].apply(map_season)
weather_final.head()
```

## Create a cross join key in both datasets

This allows every skincare product to be paired with each climate-region weather record.

# Step 5 Fix: Reduce the weather table before merging

The original merge crashed because the weather table still had too many rows.  
To fix this, we first summarize the weather data by Region and Month, then merge the smaller weather summary with the skincare dataset.

## Create a month field in the weather data

```python
weather_regions["Month"] = weather_regions["last_updated"].dt.month_name()
weather_regions.head()
```

## Create a season field

```python
def map_season(month):
    if month in ["December", "January", "February"]:
        return "Winter"
    elif month in ["March", "April", "May"]:
        return "Spring"
    elif month in ["June", "July", "August"]:
        return "Summer"
    else:
        return "Autumn"

weather_regions["Season"] = weather_regions["Month"].apply(map_season)
weather_regions.head()
```

## Summarize weather by Region and Month

This gives us a much smaller weather table for merging.

```python
weather_summary = (
    weather_regions
    .groupby(["Region", "Month", "Season"], as_index=False)
    .agg({
        "humidity": "mean",
        "uv_index": "mean",
        "temperature_celsius": "mean"
    })
)

weather_summary["humidity"] = weather_summary["humidity"].round(2)
weather_summary["uv_index"] = weather_summary["uv_index"].round(2)
weather_summary["temperature_celsius"] = weather_summary["temperature_celsius"].round(2)

print(weather_summary.shape)
weather_summary.head()
```

## Add a weather mode field

```python
weather_summary["Weather_Mode"] = "Seasonal Averages"
weather_summary.head()
```

## Create a cross join key

```python
skincare["join_key"] = 1
weather_summary["join_key"] = 1
```

## Merge the smaller weather summary with skincare

This is much lighter and should not crash the session.

```python
merged = pd.merge(skincare, weather_summary, on="join_key", how="inner")
print(merged.shape)
merged.head()
```

## Create UV Stress Level

This converts the UV index into a simple environmental stress category that will be easier to use in Tableau.

```python
def uv_stress_level(uv):
    if uv >= 8:
        return "High UV"
    elif uv >= 5:
        return "Moderate UV"
    else:
        return "Low UV"

merged["UV_Stress_Level"] = merged["uv_index"].apply(uv_stress_level)
merged[["uv_index", "UV_Stress_Level"]].head()
```

## Create Humidity Level

This groups humidity into simple moisture-condition categories for easier climate comparison.

```python
def humidity_level(h):
    if h >= 70:
        return "High Humidity"
    elif h >= 40:
        return "Moderate Humidity"
    else:
        return "Low Humidity"

merged["Humidity_Level"] = merged["humidity"].apply(humidity_level)
merged[["humidity", "Humidity_Level"]].head()
```

## Create Climate Stress Score

This combines UV and humidity conditions into one environmental stress score.
Higher UV and drier air increase the stress score.

```python
merged["Climate_Stress_Score"] = (
    merged["uv_index"] * 0.6 +
    (100 - merged["humidity"]) * 0.4 / 10
).round(2)

merged[["uv_index", "humidity", "Climate_Stress_Score"]].head()
```

## Create Adjusted Efficacy Score

This estimates product performance after accounting for ingredient sensitivity and climate stress.

```python
merged["Adjusted_Efficacy_Score"] = (
    merged["Base_Efficacy_Score"] -
    (merged["Sensitivity_Index"] * 0.8) -
    (merged["Climate_Stress_Score"] * 0.3)
).round(2)

merged[[
    "Base_Efficacy_Score",
    "Sensitivity_Index",
    "Climate_Stress_Score",
    "Adjusted_Efficacy_Score"
]].head()
```

## Create Recommended Protocol

This creates a basic treatment recommendation rule that responds to UV stress, humidity level, and ingredient type.

```python
def recommend_protocol(row):
    if row["UV_Stress_Level"] == "High UV" and row["Photosensitive_Flag"] == 1:
        return "Avoid daytime use / prioritize barrier-safe routine"
    elif row["Humidity_Level"] == "Low Humidity" and row["Moisture_Loss_Risk"] >= 2:
        return "Add hydrating support and reduce exfoliation frequency"
    elif row["Ingredient_Group"] == "Hydrators & Barrier Support":
        return "Recommended for moisture defense"
    else:
        return "Standard treatment rotation"

merged["Recommended_Protocol"] = merged.apply(recommend_protocol, axis=1)

merged[[
    "Ingredient_Group",
    "UV_Stress_Level",
    "Humidity_Level",
    "Recommended_Protocol"
]].head()
```
