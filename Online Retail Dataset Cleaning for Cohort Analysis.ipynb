# Online Retail Dataset Cleaning for Cohort Analysis

This notebook cleans the online retail dataset so it can be used for customer cohort retention analysis in Tableau.

```python
from google.colab import files
uploaded = files.upload()
```

```python
import pandas as pd
```

## Load the dataset

```python
df = pd.read_csv("online_retail.csv", encoding="ISO-8859-1")
df.head()
```

```python
print(df.shape)
print(df.columns.tolist())
```

```python
columns_needed = [
    "InvoiceNo",
    "Description",
    "Quantity",
    "InvoiceDate",
    "UnitPrice",
    "CustomerID",
    "Country"
]

df = df[columns_needed].copy()
df.head()
```

```python
df = df.dropna(subset=["CustomerID"]).copy()
print(df.shape)
```

```python
df["CustomerID"] = df["CustomerID"].astype(int)
df.head()
```

```python
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
df.dtypes
```

## Remove cancelled transactions

In this dataset, cancelled invoices usually begin with "C".
These are removed because they do not represent successful purchases.

```python
df = df[~df["InvoiceNo"].astype(str).str.startswith("C")].copy()
print(df.shape)
```

## Remove invalid quantity and price values

Only positive quantity and positive unit price values should be kept.

```python
df = df[(df["Quantity"] > 0) & (df["UnitPrice"] > 0)].copy()
print(df.shape)
```

## Create Total Amount

This is optional for the cohort dashboard, but useful for extra analysis.

```python
df["TotalAmount"] = df["Quantity"] * df["UnitPrice"]
df.head()
```

## Check the cleaned dataset

```python
print(df.shape)
print(df["CustomerID"].nunique(), "unique customers")
print(df["InvoiceDate"].min(), "to", df["InvoiceDate"].max())
```

## Save the cleaned dataset

```python
df.to_csv("online_retail_cleaned.csv", index=False)
print("Saved as online_retail_cleaned.csv")
```
