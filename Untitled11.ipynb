```python
records = []

for _, row in members.iterrows():
    member_id = row["member_id"]
    join_date = row["join_date"]

    first_stay = join_date + pd.Timedelta(days=np.random.randint(0, 60))

    if join_date.month in [11, 12, 1]:
        prob = 0.6
    elif join_date.month in [6, 7, 8]:
        prob = 0.5
    else:
        prob = 0.55

    stays = [first_stay]

    for i in range(12):
        if np.random.rand() < prob:
            stays.append(first_stay + pd.DateOffset(months=i + 1))

    for stay in stays:
        spend = round(np.random.uniform(250, 1200), 2)
        records.append([member_id, join_date, stay, spend])
```

let us now see which months of the year has the highest/lowest month_number , or member staying records. let us group our data set using their first_stay_month and then retrieve their respective month_number .This helps us to see how many members were active in their grouped month(first_stay_month)

```python
retention_counts = ( records_new.groupby(['First_Stay_Month', 'Month_Number'])["Member_ID"].unique().reset_index())
retention_counts
```

```python
cohort_s = retention_counts.iloc[:,0]

print("cohort_s:", retention_counts.head())

```

```python
retention_counts = (
    records_new.groupby(["First_Stay_Month", "Month_Number"])["Member_ID"]
    .nunique()
    .reset_index(name="Member_Count")
)

cohort_table = retention_counts.pivot(
    index="First_Stay_Month",
    columns="Month_Number",
    values="Member_Count"
)

cohort_s = cohort_table.iloc[:, 0]
cohort_table.columns = cohort_table.columns.astype(int)

retention_matrix = cohort_table.div(cohort_s, axis=0) * 100
retention_matrix = retention_matrix.round(1).fillna(0)

retention_matrix.head()
```

```python
month_number = pd.to_datetime(retention_matrix.index.astype(str)).month
```

```python
def month_class(month_num):
    if month_num in [6, 7, 8]:
        return "Peak"
    elif month_num in [11, 12, 1]:
        return "Off-Peak"
    else:
        return "Other"
```

```python
month_groups = pd.Series(month_number, index=retention_matrix.index).apply(month_class)
month_compare = retention_matrix.copy()
month_compare["Month_Groups"] = month_groups
month_compare.head()
```

the season label was added correctly , we can see it 2022-01 is marked as off-peak , while months outside the two target group are marked as otehr.

So now i need only the peak and off-peak becaus ethose are the two groups , I need for the retention comparison in the project description.

```python

peak_off_peak = month_compare[
    month_compare["Month_Groups"].isin(["Peak", "Off-Peak"])
].copy()

peak_off_peak.head()
```

Now we left with month_groups of peak or off peak only.


Now our grouped month is now labeled into month_groups based on their season , next let us start our peak offpeak comparision  
**Statistical Summary: Calculate the average 6-month retention rate for Peak vs. Off-Peak cohorts to highlight the difference in loyalty.**

```python
summary_6_month = (
    peak_off_peak.groupby("Month_Groups")[6]
    .mean()
    .reset_index()
    .rename(columns={6: "Average_6_Month_Retention"})
)

summary_6_month
```

### the 6 month retention observation:
Based on their average 6 month retention

    - off-peak : 60.55%
    - peak: 54.36%

this shows us our members who stayed in the off-peak season have high probability of retention relatice to the peak season stayed members.
