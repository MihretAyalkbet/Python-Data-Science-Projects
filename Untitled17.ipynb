```python
import pandas as pd
```

```python
df = pd.read_csv("Most-Recent-Cohorts-Institution.csv")
```

```python
df.shape
```

```python
columns_needed = [
    "UNITID",
    "INSTNM",
    "CITY",
    "STABBR",
    "ZIP",
    "CONTROL",
    "PREDDEG",
    "COSTT4_A",
    "MD_EARN_WNE_P10",
    "DEBT_MDN",
    "GRAD_DEBT_MDN",
    "PCIP11",
    "PCIP15",
    "PCIP23",
    "PCIP24",
    "PCIP38",
    "PCIP43",
    "PCIP45",
    "PCIP47",
    "PCIP48",
    "PCIP50",
    "PCIP51",
    "PCIP52",
    "PCIP54"
]

df_clean = df[columns_needed].copy()
```

```python
df_clean
```

```python
# Keep only certificate institutions
df_clean = df_clean[df_clean["PREDDEG"] == 1].copy()

# Convert numeric fields
df_clean["COSTT4_A"] = pd.to_numeric(df_clean["COSTT4_A"], errors="coerce")
df_clean["MD_EARN_WNE_P10"] = pd.to_numeric(df_clean["MD_EARN_WNE_P10"], errors="coerce")
df_clean["DEBT_MDN"] = pd.to_numeric(df_clean["DEBT_MDN"], errors="coerce")
df_clean["GRAD_DEBT_MDN"] = pd.to_numeric(df_clean["GRAD_DEBT_MDN"], errors="coerce")

# Create readable institution type
control_map = {
    1: "Public",
    2: "Private Nonprofit",
    3: "Private For-Profit"
}
df_clean["Institution_Type"] = df_clean["CONTROL"].map(control_map)

# Drop rows missing the main scatter plot values
df_clean = df_clean.dropna(subset=["COSTT4_A", "MD_EARN_WNE_P10"])

df_clean.to_csv("certificate_only_final.csv", index=False)

print(df_clean.shape)
print(df_clean["Institution_Type"].value_counts(dropna=False))
print(df_clean["PREDDEG"].value_counts())
```

```python
df_clean.to_csv("certificate_programs_cleaned.csv", index=False)
print("Saved as certificate_programs_cleaned.csv")
```
