## Check Class Proportions

The `Paint_Type` target is highly imbalanced, with Metallic dominating the dataset and Perl effect appearing very rarely.

Before modeling, I calculate the class proportions so the imbalance can be described more clearly.

```python
paint_data["Paint_Type"].value_counts(normalize=True) * 100
```

## Remove the Extremely Rare Class

The `Perl effect` category represents only a tiny fraction of the dataset, which makes it too rare for a stable train-test evaluation.

To make the classification task more reliable, I remove this class and keep the two stronger paint finish groups:
- Metallic
- Uni/basic

```python
paint_clean = paint_data[paint_data["Paint_Type"] != "Perl effect"].copy()

paint_clean["Paint_Type"].value_counts()
```

## Encode the Target Variable

Now that the dataset contains only the two main paint finish classes, I convert the target column into numeric form.

This makes it easier to train binary classification models later:
- `Metallic` → 1
- `Uni/basic` → 0

```python
paint_clean["Paint_Type"] = paint_clean["Paint_Type"].map({
    "Uni/basic": 0,
    "Metallic": 1
})

paint_clean["Paint_Type"].value_counts()
```

## Prepare the Predictor Variables

Now that the target column has been encoded, the next step is to separate the input features from the target.

At this stage, I keep `Paint_Type` as the target variable and use the remaining columns as predictor variables for the classification model.

```python
X = paint_clean.drop(columns=["Paint_Type"])
y = paint_clean["Paint_Type"]

X.head()
```

## Encode the Categorical Predictor Variables

Several of the predictor columns are stored as text, such as `make_model`, `body_type`, `Type`, `Fuel`, `Comfort_Convenience`, `Entertainment_Media`, `Safety_Security`, `Upholstery_type`, `Gearing_Type`, and `Drive_chain`.

Since machine learning models require numeric inputs, I convert the categorical predictors into numeric form using one-hot encoding.

```python
X_encoded = pd.get_dummies(X, drop_first=True)

X_encoded.head()
```

## Split the Data into Training and Testing Sets

Now that the predictor variables have been encoded, the next step is to split the data into training and testing sets.

I use stratified sampling so the class imbalance remains consistent in both the training and test sets.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

X_train.shape, X_test.shape
```

## Train the Baseline Classification Model

I begin the modeling stage with a baseline **Random Forest** classifier.

This gives me a strong starting point for the paint finish classification task and allows me to evaluate how well the model separates `Metallic` from `Uni/basic`.

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    random_state=42,
    n_estimators=100,
    class_weight="balanced"
)

rf_model.fit(X_train, y_train)

rf_predictions = rf_model.predict(X_test)
rf_probabilities = rf_model.predict_proba(X_test)[:, 1]
```

## Evaluate the Baseline Random Forest Model

Now that the Random Forest model has been trained, the next step is to evaluate its classification performance.

I begin with a classification report and confusion matrix so I can see how well the model distinguishes between `Metallic` and `Uni/basic`.

```python
from sklearn.metrics import classification_report, confusion_matrix

print(classification_report(y_test, rf_predictions))
confusion_matrix(y_test, rf_predictions)
```

The Random Forest model achieves very high overall accuracy, but the class-level results show a clear imbalance in performance. It identifies Metallic extremely well, with very high precision and recall, but performs much worse on Uni/basic, where recall is only about 0.40. The confusion matrix shows that many Uni/basic cases are being misclassified as Metallic, which means the model is biased toward the dominant class. This is important because accuracy alone makes the model look stronger than it really is for the minority finish type.

## Evaluate ROC-AUC for Random Forest

Because the dataset is highly imbalanced, I also calculate the ROC-AUC score.

This helps show how well the model separates the two paint finish classes across different probability thresholds.

```python
from sklearn.metrics import roc_auc_score

rf_roc_auc = roc_auc_score(y_test, rf_probabilities)
rf_roc_auc
```

The Random Forest model achieved a ROC-AUC of about 0.90, which indicates strong overall class separation. However, this result should be interpreted carefully because the earlier confusion matrix showed weak recall for the minority Uni/basic class. This means the model can rank the classes reasonably well overall, but it is still not handling the minority finish type as effectively as the ROC-AUC alone might suggest.

## Create the Precision-Recall Curve

Because the dataset is highly imbalanced, I also create a Precision-Recall curve.

This is especially useful here because it shows how the model behaves on the minority class and helps reveal whether strong overall performance is hiding weaker detection of `Uni/basic`.

```python
from sklearn.metrics import precision_recall_curve

precision, recall, thresholds = precision_recall_curve(y_test, rf_probabilities)

plt.figure(figsize=(8, 5))
plt.plot(recall, precision, color="darkblue")
plt.title("Precision-Recall Curve for Random Forest", fontsize=14)
plt.xlabel("Recall", fontsize=12)
plt.ylabel("Precision", fontsize=12)
plt.show()
```

The Precision-Recall curve shows that the Random Forest model maintains very high precision across most recall levels, which means its positive predictions are usually correct. However, because the minority class performed weakly in the confusion matrix, this strong curve should be interpreted carefully. It suggests that the model is confident when predicting the dominant class, but it may still struggle to recover enough true Uni/basic cases in practice.


## Summarize the Evaluation Results

To make the results easier to understand, I now create a short summary table with the main evaluation metrics.

This helps show the model’s overall accuracy, F1-score, and the most common misclassification pattern in one place.

```python
from sklearn.metrics import accuracy_score, f1_score

cm = confusion_matrix(y_test, rf_predictions)

top_misclassified_pair = "Uni/basic → Metallic" if cm[0, 1] >= cm[1, 0] else "Metallic → Uni/basic"

summary_table = pd.DataFrame({
    "Metric": ["Accuracy", "F1 Score", "ROC-AUC", "Top Misclassified Finish Pair"],
    "Value": [
        round(accuracy_score(y_test, rf_predictions), 3),
        round(f1_score(y_test, rf_predictions), 3),
        round(rf_roc_auc, 3),
        top_misclassified_pair
    ]
})

summary_table
```

## Final Interpretation and Recommendation

The Random Forest model provides strong overall performance for automotive paint finish classification, with high accuracy, a strong F1-score, and good ROC-AUC.

However, the confusion matrix and the summary table show that the model is much better at identifying the dominant `Metallic` class than the minority `Uni/basic` class. The most common error is predicting `Metallic` when the true finish is `Uni/basic`, which suggests that class imbalance is affecting the model’s reliability.

For a body shop setting, this means the model can still be useful as a support tool, but its predictions for the minority finish type should be interpreted with caution. In particular, when the model predicts `Metallic`, there is still some risk that a smaller number of `Uni/basic` cases are being absorbed into that dominant category.

Overall, this notebook shows that the classifier can provide a useful first-pass prediction for paint finish identification, but additional balancing strategies or class-sensitive modeling would likely be needed to improve minority-class detection in future versions.
