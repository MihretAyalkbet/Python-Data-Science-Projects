## Step 3: Create Battery Health Classes

To turn this into a classification problem, I now create three battery health states based on the State of Health (SOH) values.

The classes are defined as:

- **0 = Healthy**
- **1 = Degrading / Warning**
- **2 = Critical / End-of-Life**

I use SOH thresholds to create these categories, since SOH is a direct measure of battery condition and gives a clear way to separate stronger batteries from weaker ones.

```python
def classify_health(soh):
    if soh >= 80:
        return 0
    elif soh >= 60:
        return 1
    else:
        return 2

battery_data["health_class"] = battery_data["SOH"].apply(classify_health)
battery_data[["SOH", "health_class"]].head()
```

This preview shows that the health class labeling worked correctly. The first few rows all have very high SOH values, around 98 to 99, and they are all placed in class 0. That matches the logic I set, where batteries with stronger health should be labeled as Healthy. This gives me confidence that the classification rule is working the way I intended.

This is important because it shows whether the classification target is balanced or whether one class is much larger than the others.

```python
battery_data["health_class"].value_counts().sort_index()
```

The class counts show that the dataset is fairly well balanced across the three battery health categories. Class 0 has 238 rows, class 1 has 240 rows, and class 2 has 202 rows. This is a good result because it means the dataset is not heavily biased toward only one battery condition. In practical terms, the model will have enough examples from each class to learn the differences between Healthy, Degrading, and Critical battery states.

## Step 5: Prepare the Data for Classification

Now that the health classes look meaningful, the next step is to prepare the dataset for model training.

In this step, I select the main sensor and battery-life features that may help the model distinguish between Healthy, Degrading, and Critical battery states. The target variable will be the new `health_class` column.

```python
features = battery_data[[
    "cycle",
    "chI",
    "chV",
    "chT",
    "disI",
    "disV",
    "disT",
    "BCt",
    "RUL"
]]

target = battery_data["health_class"]

features.head()
```

## Split the Data into Training and Testing Sets

To evaluate the classification models fairly, I split the dataset into training and testing sets.

The models will learn from the training data and then be tested on unseen data to check how well they generalize.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    features,
    target,
    test_size=0.2,
    random_state=42,
    stratify=target
)

X_train.shape, X_test.shape
```

## Check the Class Balance After Splitting

Because this is a multiclass classification problem, I also check that the class balance is still reasonable after splitting the data.

```python
y_train.value_counts().sort_index(), y_test.value_counts().sort_index()
```

The class balance remained stable after splitting the data into training and testing sets. The training set contains 190, 192, and 162 rows across the three classes, while the test set contains 48, 48, and 40 rows. This shows that stratified splitting worked properly and preserved the overall class distribution, which is important for a fair multiclass classification evaluation.

## Step 6: Train the Classification Models

Now that the data is prepared, the next step is to train a few classification models and compare how well they can identify the three battery health states.

To follow the project requirement, I begin with three different models:

- Logistic Regression
- Random Forest Classifier
- Gradient Boosting Classifier

This gives me a baseline comparison between a simple linear model and two more flexible tree-based models.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
```

## Step 6: Train the First Classification Model

Now that the dataset is prepared, I begin the modeling stage with a first baseline classifier.

I start with **Logistic Regression** because it is a simple and interpretable model that can give a useful baseline for the battery health classification task.

```python
from sklearn.linear_model import LogisticRegression
```

## Train Logistic Regression

Here, I train the Logistic Regression model using the training data.

```python
log_model = LogisticRegression(max_iter=1000, random_state=42)
log_model.fit(X_train, y_train)

log_predictions = log_model.predict(X_test)
```

## Fix the Logistic Regression Warning

The first Logistic Regression run produced a convergence warning.

This usually happens when the input features are on different scales. To improve the training process, I now standardize the features and retrain the model.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
```

## Retrain Logistic Regression with Scaled Features

Here, I use a scaling step together with Logistic Regression so the model can train more smoothly.

```python
log_model = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=2000, random_state=42)
)

log_model.fit(X_train, y_train)

log_predictions = log_model.predict(X_test)
```

## Step 7: Evaluate Logistic Regression

Now that the Logistic Regression model has trained successfully, the next step is to evaluate how well it predicts the three battery health classes.

I begin with a classification report, which shows precision, recall, and F1-score for each class.

```python
from sklearn.metrics import classification_report

print(classification_report(y_test, log_predictions))
```

The Logistic Regression model performed very well across all three battery health classes. It achieved an overall accuracy of 0.96, and the weighted F1-score was also 0.96. The Healthy class was predicted especially well, with an F1-score of 0.99, while both the Degrading and Critical classes reached strong F1-scores of 0.95. This suggests that the model is doing a very good job separating the three battery conditions.

## Confusion Matrix for Logistic Regression

To better understand where the model is making correct and incorrect predictions, I also visualize the confusion matrix.

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
import matplotlib.pyplot as plt

cm = confusion_matrix(y_test, log_predictions)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Healthy", "Degrading", "Critical"]
)

disp.plot(cmap="Blues")
plt.title("Confusion Matrix - Logistic Regression")
plt.show()
```

The confusion matrix shows that most battery cycles were classified correctly. The model predicted 47 out of 48 Healthy cases correctly, 45 out of 48 Degrading cases correctly, and 39 out of 40 Critical cases correctly. Most of the mistakes happened between the Degrading and Critical classes, which makes sense because those two conditions are more similar than Healthy batteries. Overall, the confusion matrix confirms that Logistic Regression is performing strongly and is especially reliable at detecting the Critical class.

## Step 8: Train the Second Classification Model

Now that the Logistic Regression model has been evaluated, I move to the second classifier.

I use a **Random Forest Classifier** next, because it can capture more complex non-linear patterns in the battery features and may improve the separation between the health classes.

```python
from sklearn.ensemble import RandomForestClassifier
```

## Train Random Forest Classifier

Here, I train the Random Forest model using the same training and testing split.

```python
rf_model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf_model.fit(X_train, y_train)

rf_predictions = rf_model.predict(X_test)
```

## Step 9: Evaluate the Random Forest Model

Now that the Random Forest model has been trained, the next step is to evaluate how well it predicts the three battery health classes.

I begin with the classification report so I can compare its precision, recall, and F1-scores with the Logistic Regression model.

```python
print(classification_report(y_test, rf_predictions))
```

The Random Forest model performed perfectly on the test set. It achieved 1.00 for precision, recall, and F1-score across all three battery health classes, with an overall accuracy of 1.00. This means the model was able to separate Healthy, Degrading, and Critical battery states without making any classification errors on the test data.

## Confusion Matrix for Random Forest

To better understand how the Random Forest model is making correct and incorrect predictions, I also visualize its confusion matrix.

```python
cm_rf = confusion_matrix(y_test, rf_predictions)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_rf,
    display_labels=["Healthy", "Degrading", "Critical"]
)

disp.plot(cmap="Greens")
plt.title("Confusion Matrix - Random Forest")
plt.show()
```

The confusion matrix confirms that the Random Forest model classified every test case correctly. All 48 Healthy cases, all 48 Degrading cases, and all 40 Critical cases were placed into their correct classes, with no misclassifications at all. This suggests that the Random Forest model is capturing the battery health patterns extremely well and is performing better than Logistic Regression.

## Step 10: Train the Third Classification Model

After evaluating Logistic Regression and Random Forest, I now move to the third classifier.

I use **Gradient Boosting Classifier** next, because it is another strong tree-based model that often performs well on structured tabular data and can give a useful comparison against Random Forest.

```python
from sklearn.ensemble import GradientBoostingClassifier
```

## Train Gradient Boosting Classifier

Here, I train the Gradient Boosting model using the same training and testing split.

```python
gb_model = GradientBoostingClassifier(random_state=42)
gb_model.fit(X_train, y_train)

gb_predictions = gb_model.predict(X_test)
```

## Step 11: Evaluate the Gradient Boosting Model

Now that the Gradient Boosting model has been trained, the next step is to evaluate how well it predicts the three battery health classes.

I begin with the classification report so I can compare its precision, recall, and F1-scores with the previous models.

```python
print(classification_report(y_test, gb_predictions))
```

The Gradient Boosting model also performed very strongly on the test set. It achieved an overall accuracy of 0.99, with very high precision, recall, and F1-scores across all three classes. The Critical class was classified especially well, with perfect precision, recall, and F1-score, while the Healthy and Degrading classes also showed nearly perfect results. This means Gradient Boosting is a very strong model, although it still performs slightly below the Random Forest model.

## Confusion Matrix for Gradient Boosting

To understand how the Gradient Boosting model is making correct and incorrect predictions, I also visualize its confusion matrix.

```python
cm_gb = confusion_matrix(y_test, gb_predictions)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_gb,
    display_labels=["Healthy", "Degrading", "Critical"]
)

disp.plot(cmap="Oranges")
plt.title("Confusion Matrix - Gradient Boosting")
plt.show()
```

The confusion matrix shows that the Gradient Boosting model made only one mistake on the test set. It correctly classified 47 out of 48 Healthy cases, all 48 Degrading cases, and all 40 Critical cases. The only error was one Healthy battery being predicted as Degrading. Overall, this confirms that the model performs extremely well and is especially reliable at identifying the Critical class.

## Step 12: Compare the Three Models

Now that all three models have been trained and evaluated, the next step is to compare them directly.

This helps me identify which model performs best overall and which one is most reliable for detecting the Critical battery class, which is the most important class for maintenance decisions.

```python
from sklearn.metrics import f1_score, accuracy_score

model_comparison = pd.DataFrame({
    "Model": ["Logistic Regression", "Random Forest", "Gradient Boosting"],
    "Accuracy": [
        accuracy_score(y_test, log_predictions),
        accuracy_score(y_test, rf_predictions),
        accuracy_score(y_test, gb_predictions)
    ],
    "Weighted F1": [
        f1_score(y_test, log_predictions, average="weighted"),
        f1_score(y_test, rf_predictions, average="weighted"),
        f1_score(y_test, gb_predictions, average="weighted")
    ],
    "Critical Class F1": [
        f1_score(y_test, log_predictions, labels=[2], average="macro"),
        f1_score(y_test, rf_predictions, labels=[2], average="macro"),
        f1_score(y_test, gb_predictions, labels=[2], average="macro")
    ]
})

model_comparison.round(3)
```

## Interpret the Model Comparison

This comparison table helps summarize the strengths of each model.

The most important measure here is the performance on the **Critical** class, because that class represents the highest battery risk and is the most important one for proactive maintenance decisions.

The model comparison table shows that all three classifiers performed strongly, but Random Forest was the best overall. It achieved perfect accuracy, weighted F1, and Critical Class F1, which means it classified every test example correctly, including the most important Critical class. Gradient Boosting also performed very well and reached a perfect F1-score for the Critical class, but its overall performance was slightly lower than Random Forest. Logistic Regression still performed strongly, but it was the weakest of the three models. Based on these results, Random Forest is the most reliable model for battery health classification and is the best choice for proactive maintenance decisions.
