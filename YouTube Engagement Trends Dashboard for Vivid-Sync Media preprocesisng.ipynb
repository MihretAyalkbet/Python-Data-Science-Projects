# Step 2: Clean the dataset and create Tableau-ready time fields

In this step, we clean the core YouTube fields, convert the date columns, ensure numeric columns are valid, and create helper fields for time-based Tableau analysis.

## Keep only the fields needed for the project

```python
columns_needed = [
    "video_id",
    "trending_date",
    "title",
    "channel_title",
    "category_id",
    "publish_time",
    "views",
    "likes",
    "comment_count",
    "Region"
]

df = df[columns_needed].copy()
df.head()
```

## Remove rows with missing values in the core fields

```python
df = df.dropna(subset=[
    "trending_date",
    "title",
    "channel_title",
    "category_id",
    "publish_time",
    "views",
    "likes",
    "comment_count"
]).copy()

print(df.shape)
```

## Convert the numeric fields to numeric type

```python
for col in ["views", "likes", "comment_count", "category_id"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print(df[["views", "likes", "comment_count", "category_id"]].dtypes)
```

## Remove rows that became invalid after numeric conversion

```python
df = df.dropna(subset=["views", "likes", "comment_count", "category_id"]).copy()
print(df.shape)
```

## Convert publish time to datetime

```python
df["publish_time"] = pd.to_datetime(df["publish_time"], errors="coerce")
print(df["publish_time"].dtype)
```

## Convert trending date to datetime

The trending_date field in this dataset uses a short year format like YY.DD.MM.

```python
df["trending_date"] = pd.to_datetime(df["trending_date"], format="%y.%d.%m", errors="coerce")
print(df["trending_date"].dtype)
```

## Remove rows with invalid converted dates

```python
df = df.dropna(subset=["publish_time", "trending_date"]).copy()
print(df.shape)
```

## Create helper date fields for Tableau

```python
df["Trending_Month"] = df["trending_date"].dt.to_period("M").astype(str)
df["Trending_Week"] = df["trending_date"].dt.to_period("W").astype(str)
df["Trending_Year"] = df["trending_date"].dt.year

df[["trending_date", "Trending_Month", "Trending_Week", "Trending_Year"]].head()
```

## Preview the cleaned dataset

```python
print(df.shape)
df.head()
```

# Step 3: Create category labels and save the Tableau-ready dataset

In this step, we map the YouTube category IDs to readable category names, create a cleaner category field for Tableau, and save the final dataset.

## Map category IDs to readable YouTube category names

```python
category_map = {
    1: "Film & Animation",
    2: "Autos & Vehicles",
    10: "Music",
    15: "Pets & Animals",
    17: "Sports",
    18: "Short Movies",
    19: "Travel & Events",
    20: "Gaming",
    21: "Videoblogging",
    22: "People & Blogs",
    23: "Comedy",
    24: "Entertainment",
    25: "News & Politics",
    26: "Howto & Style",
    27: "Education",
    28: "Science & Technology",
    29: "Nonprofits & Activism",
    30: "Movies",
    31: "Anime/Animation",
    32: "Action/Adventure",
    33: "Classics",
    34: "Comedy",
    35: "Documentary",
    36: "Drama",
    37: "Family",
    38: "Foreign",
    39: "Horror",
    40: "Sci-Fi/Fantasy",
    41: "Thriller",
    42: "Shorts",
    43: "Shows",
    44: "Trailers"
}

df["Category"] = df["category_id"].map(category_map).fillna("Other")
df[["category_id", "Category"]].drop_duplicates().sort_values("category_id").head(20)
```

```python
df["Category"].value_counts().head(20)
```

## Standardize text fields

```python
for col in ["title", "channel_title", "Category", "Region"]:
    df[col] = df[col].astype(str).str.strip()

df.head()
```
