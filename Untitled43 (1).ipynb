# Model Evaluation Strategy for Sleep Apnea Screening Triage

## Business Problem

Verdant Pulse Sleep Clinic wants to use a data-driven screening process to help prioritize patients for overnight sleep studies.

The main goal of this project is to compare several classification models and determine which one is safest to use for triage. In this setting, missing a patient with sleep apnea is more serious than sending a healthy patient for extra testing, so the evaluation must focus on clinical risk and not only on overall accuracy.

## Project Objective

The objective of this project is to:

- load and understand the sleep health dataset
- explore how biometric variables relate to sleep disorder categories
- train at least three classification models
- compare their performance using clinical safety metrics
- analyze false negatives and false positives
- identify which features matter most
- recommend the most appropriate model for triage use

## Step 2: Clean the Target Column and Check Missing Values

From the dataset overview, the `Sleep Disorder` column has missing values, while the other columns appear complete.

Since this project depends on classifying sleep disorder categories, I first need to inspect the target column more closely and remove rows where the target is missing.

```python
sleep_data["Sleep Disorder"].value_counts(dropna=False)
```

## Remove Rows with Missing Target Values

The `Sleep Disorder` column contains many missing values. Since this project is a supervised classification task, rows without a target label cannot be used for model training.

So in this step, I remove the rows where `Sleep Disorder` is missing.

```python
sleep_clean = sleep_data.dropna(subset=["Sleep Disorder"]).copy()

sleep_clean.shape
```

## Check the Class Balance After Cleaning

Now that the missing target values have been removed, I check the class balance again.

This helps me confirm how many observations remain in each sleep disorder category before moving to feature cleaning and modeling.

```python
sleep_clean["Sleep Disorder"].value_counts()
```

## Check for Missing Values in the Remaining Features

Now that the target column has been cleaned, I check whether the other predictor columns still contain missing values.

This helps confirm whether the rest of the dataset is ready for preprocessing and modeling.

```python
sleep_clean.isnull().sum()
```

## Remove Unnecessary Identifier Column

The dataset no longer has missing values, so the next step is to prepare the features for modeling.

Since `Person ID` is only an identifier and does not contain meaningful clinical information, I remove it before preprocessing the remaining variables.

```python
sleep_clean = sleep_clean.drop(columns=["Person ID"])

sleep_clean.head()
```

## Encode the Target Variable

The target column is still stored as text labels, so the next step is to convert it into numeric form.

This makes it easier to train classification models later while still preserving the meaning of the sleep disorder categories.

```python
sleep_clean["Sleep Disorder"] = sleep_clean["Sleep Disorder"].map({
    "Insomnia": 0,
    "Sleep Apnea": 1
})

sleep_clean["Sleep Disorder"].value_counts()
```

## Prepare the Predictor Variables

Now that the target variable has been encoded, the next step is to separate the predictor variables from the target.

At this stage, I keep all remaining columns except `Sleep Disorder` as input features.

```python
X = sleep_clean.drop(columns=["Sleep Disorder"])
y = sleep_clean["Sleep Disorder"]

X.head()
```

## Encode the Categorical Predictor Variables

Some of the predictor columns are still stored as text, such as `Gender`, `Occupation`, `BMI Category`, and `Blood Pressure`.

Since machine learning models need numeric input, I now convert the categorical variables into numeric form using one-hot encoding.

```python
X_encoded = pd.get_dummies(X, drop_first=True)

X_encoded.head()
```

## Split the Data into Training and Testing Sets

Now that the predictor variables have been encoded, the next step is to split the data into training and testing sets.

This allows me to train the models on one part of the dataset and evaluate them fairly on unseen data.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

X_train.shape, X_test.shape
```

##  Train the First Classification Model

I begin the modeling stage with a baseline **Logistic Regression** model.

This gives me a simple starting point that I can later compare with more flexible models such as Random Forest and Gradient Boosting.

```python
from sklearn.linear_model import LogisticRegression

log_model = LogisticRegression(max_iter=5000, random_state=42)
log_model.fit(X_train, y_train)

log_predictions = log_model.predict(X_test)
log_probabilities = log_model.predict_proba(X_test)[:, 1]
```

## Scale the Features for Logistic Regression

The convergence warning is still appearing, which means the model is having trouble optimizing with the current feature scales.

To fix this, I standardize the encoded feature values before fitting Logistic Regression again.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## Retrain Logistic Regression with Scaled Features

Now I train the Logistic Regression model again using the scaled data.

```python
log_model = LogisticRegression(max_iter=5000, random_state=42)
log_model.fit(X_train_scaled, y_train)

log_predictions = log_model.predict(X_test_scaled)
log_probabilities = log_model.predict_proba(X_test_scaled)[:, 1]
```

## Evaluate the Baseline Logistic Regression Model

Now that the Logistic Regression model has trained successfully, the next step is to evaluate its baseline performance.

I start with a classification report and confusion matrix so I can see how well the model identifies Sleep Apnea versus Insomnia.

```python
from sklearn.metrics import classification_report, confusion_matrix

print(classification_report(y_test, log_predictions))
confusion_matrix(y_test, log_predictions)
```

The baseline Logistic Regression model performs well, with an overall accuracy of about 87% and balanced precision, recall, and F1-scores across both classes. The confusion matrix shows only a small number of misclassifications, with 2 false negatives and 2 false positives. This suggests that Logistic Regression provides a strong starting point for the sleep apnea triage task, although the false negatives remain important because missing a patient with apnea is clinically more costly than over-testing a healthy patient.

## Evaluate ROC-AUC for Logistic Regression

Accuracy and F1-score are useful, but for this project I also need a probability-based measure of model quality.

ROC-AUC helps show how well the Logistic Regression model separates the two classes across different classification thresholds.

```python
from sklearn.metrics import roc_auc_score

log_roc_auc = roc_auc_score(y_test, log_probabilities)
log_roc_auc
```

The Logistic Regression model achieved a ROC-AUC of about 0.94, which indicates very strong class separation. This means the model is highly capable of distinguishing between Sleep Apnea and Insomnia across different decision thresholds. From a triage perspective, this is a strong result because it suggests the model has useful ranking ability even before adjusting the classification threshold for clinical safety.

## Create the Precision-Recall Curve for Logistic Regression

Because this project is focused on clinical safety, I also check the Precision-Recall trade-off.

This is important because in the sleep apnea triage setting, recall matters a lot. Missing a patient with apnea is more costly than sending an extra patient for testing, so I want to see how precision changes as recall changes.

```python
from sklearn.metrics import precision_recall_curve

precision, recall, thresholds = precision_recall_curve(y_test, log_probabilities)

plt.figure(figsize=(8, 5))
plt.plot(recall, precision, color="darkred")
plt.title("Precision-Recall Curve for Logistic Regression", fontsize=14)
plt.xlabel("Recall", fontsize=12)
plt.ylabel("Precision", fontsize=12)
plt.show()
```

The precision-recall curve shows that the Logistic Regression model maintains very strong precision across a wide range of recall values. Precision stays close to 1.0 until recall becomes quite high, which suggests the model can identify many apnea cases without introducing many false positives. This is a strong result for triage, because it means the model has the potential to support high-recall screening while still keeping precision at a useful level.

## Train the Second Classification Model

After evaluating Logistic Regression, the next step is to train a more flexible model for comparison.

I now train a Random Forest classifier, which can capture non-linear relationships and interactions between biometric features.

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(random_state=42, n_estimators=100)
rf_model.fit(X_train, y_train)

rf_predictions = rf_model.predict(X_test)
rf_probabilities = rf_model.predict_proba(X_test)[:, 1]
```

## Evaluate the Random Forest Model

Now that the Random Forest model has been trained, the next step is to evaluate its performance.

I begin with a classification report and confusion matrix so I can compare it directly with the Logistic Regression model.

```python
print(classification_report(y_test, rf_predictions))
confusion_matrix(y_test, rf_predictions)
```

## Evaluate ROC-AUC for Random Forest

To compare the Random Forest model more fairly with Logistic Regression, I now calculate the ROC-AUC score.

This helps show how well the Random Forest model separates the two classes across different probability thresholds.

```python
rf_roc_auc = roc_auc_score(y_test, rf_probabilities)
rf_roc_auc
```

The Random Forest model achieved a ROC-AUC of about 0.94, which is also a very strong result. This means the model separates Sleep Apnea and Insomnia very well across different probability thresholds. Its ROC-AUC is almost the same as Logistic Regression, so at this stage both models appear similarly strong in their overall class separation ability.

## Create the Precision-Recall Curve for Random Forest

Because this project emphasizes clinical safety, I also check the Precision-Recall trade-off for the Random Forest model.

This helps show whether the model can maintain strong precision while also capturing a high proportion of apnea cases.

```python
precision_rf, recall_rf, thresholds_rf = precision_recall_curve(y_test, rf_probabilities)

plt.figure(figsize=(8, 5))
plt.plot(recall_rf, precision_rf, color="darkgreen")
plt.title("Precision-Recall Curve for Random Forest", fontsize=14)
plt.xlabel("Recall", fontsize=12)
plt.ylabel("Precision", fontsize=12)
plt.show()
```

The Random Forest precision-recall curve shows strong performance, with precision remaining high across most recall levels. This suggests the model can identify a large share of sleep apnea cases without creating too many false positives. Compared with Logistic Regression, the overall pattern looks very similar, which means both models currently offer strong clinical screening potential.

## Train the Third Classification Model

To complete the comparative modeling requirement, I now train a third classifier.

I use Gradient Boosting because it is another strong ensemble method that can capture complex patterns in the biometric data.

```python
from sklearn.ensemble import GradientBoostingClassifier

gb_model = GradientBoostingClassifier(random_state=42)
gb_model.fit(X_train, y_train)

gb_predictions = gb_model.predict(X_test)
gb_probabilities = gb_model.predict_proba(X_test)[:, 1]
```

## Evaluate the Gradient Boosting Model

Now that the Gradient Boosting model has been trained, the next step is to evaluate its classification performance.

I begin with a classification report and confusion matrix so I can compare it directly with Logistic Regression and Random Forest.

```python
print(classification_report(y_test, gb_predictions))
confusion_matrix(y_test, gb_predictions)
```

The Gradient Boosting model performs almost identically to the other two models. It achieved about 87% accuracy, with balanced precision, recall, and F1-scores across both classes. The confusion matrix again shows 2 false negatives and 2 false positives, which means its clinical error pattern is currently the same as Logistic Regression and Random Forest.

## Evaluate ROC-AUC for Gradient Boosting

To complete the model comparison fairly, I now calculate the ROC-AUC score for the Gradient Boosting model.

This helps show how well the model separates the two classes across different probability thresholds.

```python
gb_roc_auc = roc_auc_score(y_test, gb_probabilities)
gb_roc_auc
```

The Gradient Boosting model achieved a ROC-AUC of about 0.91, which is still a strong result, but slightly lower than the ROC-AUC values for Logistic Regression and Random Forest. This means Gradient Boosting can separate Sleep Apnea and Insomnia well, but at this stage it appears a little less effective than the other two models in overall class discrimination.

## Create the Precision-Recall Curve for Gradient Boosting

To complete the probability-based evaluation, I now create the Precision-Recall curve for the Gradient Boosting model.

This is useful because the project focuses on clinical safety, and recall is especially important when missing a patient with sleep apnea is costly.

```python
precision_gb, recall_gb, thresholds_gb = precision_recall_curve(y_test, gb_probabilities)

plt.figure(figsize=(8, 5))
plt.plot(recall_gb, precision_gb, color="purple")
plt.title("Precision-Recall Curve for Gradient Boosting", fontsize=14)
plt.xlabel("Recall", fontsize=12)
plt.ylabel("Precision", fontsize=12)
plt.show()
```

The Precision-Recall curve for Gradient Boosting shows strong performance, with precision staying high across most recall values. However, compared with Logistic Regression and Random Forest, the curve drops a little earlier, which suggests Gradient Boosting may be slightly weaker in maintaining precision at higher recall levels. This means it is still a useful model, but it appears a bit less reliable than the other two for a safety-focused triage setting.

## Compare the Three Models in One Summary Table

Now that all three models have been evaluated, the next step is to compare them in one table.

This makes it easier to see which model performs best across accuracy, F1-score, and ROC-AUC before making a final recommendation.

```python
from sklearn.metrics import accuracy_score, f1_score

model_summary = pd.DataFrame({
    "Model": ["Logistic Regression", "Random Forest", "Gradient Boosting"],
    "Accuracy": [
        accuracy_score(y_test, log_predictions),
        accuracy_score(y_test, rf_predictions),
        accuracy_score(y_test, gb_predictions)
    ],
    "F1 Score": [
        f1_score(y_test, log_predictions),
        f1_score(y_test, rf_predictions),
        f1_score(y_test, gb_predictions)
    ],
    "ROC-AUC": [
        log_roc_auc,
        rf_roc_auc,
        gb_roc_auc
    ]
})

model_summary.round(3)
```

The model comparison table shows that all three models achieved the same accuracy and F1-score, which means they performed equally well at the default classification threshold. The main difference appears in ROC-AUC, where Logistic Regression scored the highest at 0.942, followed closely by Random Forest at 0.938, while Gradient Boosting performed slightly lower at 0.910. This suggests that Logistic Regression provides the strongest overall class separation and is currently the best candidate for a safety-focused triage model.

## Clinical Risk Assessment Using the Confusion Matrix

Because this project is focused on clinical triage, I now evaluate the models from a safety perspective.

In this setting, a **false negative** is more costly than a **false positive**, because missing a patient with sleep apnea is more serious than sending a healthy patient for extra testing.

To reflect that, I create a simple cost-based comparison using the confusion matrix for each model.

```python
log_cm = confusion_matrix(y_test, log_predictions)
rf_cm = confusion_matrix(y_test, rf_predictions)
gb_cm = confusion_matrix(y_test, gb_predictions)

def calculate_clinical_cost(cm, fn_cost=5, fp_cost=1):
    tn, fp, fn, tp = cm.ravel()
    return (fn * fn_cost) + (fp * fp_cost)

clinical_risk_summary = pd.DataFrame({
    "Model": ["Logistic Regression", "Random Forest", "Gradient Boosting"],
    "False Positives": [log_cm[0, 1], rf_cm[0, 1], gb_cm[0, 1]],
    "False Negatives": [log_cm[1, 0], rf_cm[1, 0], gb_cm[1, 0]],
    "Clinical Cost Score": [
        calculate_clinical_cost(log_cm),
        calculate_clinical_cost(rf_cm),
        calculate_clinical_cost(gb_cm)
    ]
})

clinical_risk_summary
```

The clinical risk summary shows that all three models produce the same error pattern: 2 false positives and 2 false negatives, leading to the same clinical cost score of 12. This means none of the models has an advantage over the others when evaluated purely from the clinic’s false-negative versus false-positive risk perspective. Since the clinical cost is tied, the final recommendation should depend on the stronger probability-based performance, where Logistic Regression has the highest ROC-AUC.

## Feature Importance and Interpretability

The final step is to examine which predictors are driving the model decisions.

Since Logistic Regression has the strongest ROC-AUC and performs equally well on the clinical cost analysis, I use it as the main model for interpretation.

```python
feature_importance = pd.DataFrame({
    "Feature": X_encoded.columns,
    "Coefficient": log_model.coef_[0]
})

feature_importance["Absolute_Coefficient"] = feature_importance["Coefficient"].abs()

feature_importance = feature_importance.sort_values(
    by="Absolute_Coefficient",
    ascending=False
)

feature_importance.head(10)
```

The feature importance results suggest that the Logistic Regression model is relying most strongly on a combination of occupation, blood pressure, heart rate, and sleep-related measures when distinguishing between Sleep Apnea and Insomnia. In particular, variables such as Occupation_Nurse, Blood_Pressure_140/95, Heart Rate, and Sleep Duration have some of the largest coefficients in absolute value, which means they are contributing strongly to the model’s decision boundary. This supports the idea that both physiological indicators and lifestyle-related factors play an important role in the triage model.
