## Correct the Date Conversion

## Rebuild the Weather Table with the Correct Date Conversion

Because the earlier date conversion changed the values into an incorrect timestamp format, I now rebuild the Heathrow weather table from the original weather dataset and convert the date column again using the correct `YYYYMMDD` format.

```python
heathrow_weather = weather_data[[
    "DATE",
    "HEATHROW_temp_mean",
    "HEATHROW_temp_min",
    "HEATHROW_temp_max"
]].copy()

heathrow_weather = heathrow_weather.rename(columns={
    "DATE": "date",
    "HEATHROW_temp_mean": "temp_mean",
    "HEATHROW_temp_min": "temp_min",
    "HEATHROW_temp_max": "temp_max"
})

heathrow_weather["date"] = pd.to_datetime(
    heathrow_weather["date"].astype(str),
    format="%Y%m%d"
)

heathrow_weather.head()
```

## Step 4: Create a Representative Booking Dataset

Now that the weather data is prepared, the next step is to create a representative booking dataset for the wellness studio.

Since the project allows the booking data to be generated by script, I create two booking categories:

- **Heat** sessions
- **Ice** sessions

The idea is to simulate realistic demand patterns where warmer days tend to encourage more Ice bookings, while colder days tend to support more Heat bookings.

```python
booking_data = heathrow_weather[["date", "temp_mean"]].copy()

np.random.seed(42)

booking_data["heat_bookings"] = (
    40
    - 0.8 * booking_data["temp_mean"]
    + np.random.normal(0, 3, len(booking_data))
).round()

booking_data["ice_bookings"] = (
    10
    + 0.9 * booking_data["temp_mean"]
    + np.random.normal(0, 3, len(booking_data))
).round()

booking_data["heat_bookings"] = booking_data["heat_bookings"].clip(lower=0)
booking_data["ice_bookings"] = booking_data["ice_bookings"].clip(lower=0)

booking_data.head()
```

## Step 5: Reshape the Booking Data for Comparison

Right now, the booking dataset stores Heat and Ice bookings in separate columns.

To make the later analysis and visualization easier, I now reshape the data into a longer format so each row represents one booking category on one date.

```python
booking_long = booking_data.melt(
    id_vars=["date", "temp_mean"],
    value_vars=["heat_bookings", "ice_bookings"],
    var_name="session_type",
    value_name="bookings"
)

booking_long["session_type"] = booking_long["session_type"].replace({
    "heat_bookings": "Heat",
    "ice_bookings": "Ice"
})

booking_long.head()
```

The booking data was reshaped successfully into a long format. Each row now represents one session type on one date, with its corresponding temperature and booking count. This structure is much better for comparison because it will make it easier to analyze Heat and Ice bookings side by side in the next visual steps.

## Step 6: Visualize Booking Trends by Temperature

Now that the booking data is in a comparison-friendly format, the next step is to visualize how bookings change with temperature.

This scatter plot helps show whether Heat and Ice sessions respond differently to outdoor temperature.

```python
plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=booking_long,
    x="temp_mean",
    y="bookings",
    hue="session_type"
)
plt.title("Bookings by Temperature and Session Type", fontsize=14)
plt.xlabel("Average Outdoor Temperature", fontsize=12)
plt.ylabel("Bookings", fontsize=12)
plt.show()
```

The scatter plot shows a very clear opposite relationship between temperature and the two session types. Heat bookings tend to decrease as outdoor temperature rises, while Ice bookings increase as the weather gets warmer. The two patterns move toward each other and begin to overlap in the middle temperature range, which suggests there may be a pivot point where customer preference shifts from Heat-dominant to Ice-dominant bookings.

## Step 7: Compare Average Bookings Across Temperature Brackets

To make the booking pattern easier to interpret, I now group the temperature values into brackets and compare the average number of bookings for Heat and Ice sessions within each bracket.

This will help show more clearly where the shift in customer preference begins to happen.

```python
booking_long["temp_bracket"] = pd.cut(
    booking_long["temp_mean"],
    bins=[-5, 0, 5, 10, 15, 20, 25, 30],
    labels=["<0", "0-5", "5-10", "10-15", "15-20", "20-25", "25-30"]
)

bracket_summary = (
    booking_long.groupby(["temp_bracket", "session_type"], observed=False)["bookings"]
    .mean()
    .reset_index()
)

bracket_summary.head()
```

The temperature bracket summary already shows a strong difference between the two session types at lower temperatures. In the coldest bracket, Heat bookings are much higher than Ice bookings, and this same pattern continues in the early cool ranges. This supports the idea that colder weather favors Heat sessions and suggests that the booking balance will likely shift as the temperature brackets move upward.

## Step 8: Visualize Average Bookings by Temperature Bracket

Now I turn the grouped summary into a chart so the booking shift is easier to see.

This bar chart compares the average Heat and Ice bookings across each temperature bracket and helps show where the booking preference begins to flip.

```python
plt.figure(figsize=(10, 5))
sns.barplot(
    data=bracket_summary,
    x="temp_bracket",
    y="bookings",
    hue="session_type"
)

plt.title("Average Bookings by Temperature Bracket", fontsize=14)
plt.xlabel("Temperature Bracket", fontsize=12)
plt.ylabel("Average Bookings", fontsize=12)
plt.show()
```

The bar chart makes the booking pattern much clearer. Heat sessions dominate strongly at lower temperatures, while Ice sessions grow steadily as the weather gets warmer. Around the 15–20°C bracket, the two booking types become nearly equal, and after that point Ice bookings begin to overtake Heat bookings. This suggests that the pivot point is somewhere around the mid-to-high teen temperatures, where customer preference starts shifting from Heat-dominant to Ice-dominant demand.

## Step 9: Estimate the Pivot Temperature

Now that the bracket chart shows where the balance begins to shift, the next step is to estimate the pivot temperature more directly.

I do this by comparing the daily difference between Heat and Ice bookings. The pivot point is the temperature where that difference gets close to zero, meaning the two session types are nearly equally preferred.

```python
booking_data["booking_gap"] = booking_data["heat_bookings"] - booking_data["ice_bookings"]
booking_data[["date", "temp_mean", "heat_bookings", "ice_bookings", "booking_gap"]].head()
```

The booking gap column was created successfully. A positive gap means Heat bookings are higher than Ice bookings, while a negative gap would mean Ice bookings are higher. In these early rows, the gap is strongly positive, which matches the cooler temperatures shown and confirms that Heat sessions are still dominating at this stage of the data.

## Step 10: Find the Days Where the Booking Gap Is Closest to Zero

To estimate the pivot point more directly, I now look for the days where the booking gap is closest to zero.

These are the days where Heat and Ice bookings are most balanced, which gives a practical estimate of the outdoor temperature where customer preference begins to flip.

```python
pivot_candidates = booking_data.iloc[
    booking_data["booking_gap"].abs().argsort()
][["date", "temp_mean", "heat_bookings", "ice_bookings", "booking_gap"]].head(10)

pivot_candidates.tail(1000)
```

## Step 11: Estimate the Pivot Temperature Range

Now that I have identified the days where Heat and Ice bookings are balanced, the next step is to estimate the pivot temperature range.

I do this by summarizing the temperatures from the pivot candidate days. This gives a practical estimate of the outdoor temperature where booking preference begins to switch.

```python
pivot_candidates["temp_mean"].describe()
```

The pivot candidate summary suggests that the shift in customer preference happens in the mid-teen temperature range. The average balanced temperature is about 16.4°C, and the middle half of the pivot days falls roughly between 15.4°C and 17.9°C. This gives a practical estimate that the booking preference begins to flip at around 16°C to 17°C.

## Step 12: Visualize the Booking Gap Against Temperature

To make the pivot point easier to interpret, I now plot the booking gap directly against temperature.

This chart helps show where the booking gap crosses zero, which makes the shift from Heat-dominant demand to Ice-dominant demand easier to see.

```python
plt.figure(figsize=(9, 5))
sns.scatterplot(data=booking_data, x="temp_mean", y="booking_gap", color="purple")
plt.axhline(0, color="black", linestyle="--")
plt.title("Booking Gap vs Temperature", fontsize=14)
plt.xlabel("Average Outdoor Temperature", fontsize=12)
plt.ylabel("Heat Bookings - Ice Bookings", fontsize=12)
plt.show()
```

The booking gap plot makes the pivot point very clear. At lower temperatures, the gap is strongly positive, which means Heat bookings are much higher than Ice bookings. As temperature rises, the gap steadily drops, and around the mid-teen temperatures it crosses the zero line. After that point, the gap becomes negative, which shows that Ice bookings begin to overtake Heat bookings. This confirms a strong temperature-driven shift in customer preference.

## Step 13: Create an Interactive Version of the Pivot Analysis

To make the temperature-driven booking shift easier to explore, I now create an interactive scatter plot.

This allows the analyst to hover over points and inspect the exact temperature, booking counts, and booking gap on specific days.

```python
import plotly.express as px

fig = px.scatter(
    booking_data,
    x="temp_mean",
    y="booking_gap",
    hover_data=["date", "heat_bookings", "ice_bookings"],
    title="Interactive Booking Gap vs Temperature",
    labels={
        "temp_mean": "Average Outdoor Temperature",
        "booking_gap": "Heat Bookings - Ice Bookings"
    },
    color="booking_gap",
    color_continuous_scale="RdBu"
)

fig.add_hline(y=0, line_dash="dash", line_color="black")
fig.show()
```

## Step 14: Summarize the Pivot Findings

Now that the pivot pattern has been shown in both static and interactive form, the next step is to summarize the main result in a compact table.

This helps turn the visual analysis into a more direct business-ready finding.

```python
pivot_summary = pd.DataFrame({
    "Metric": [
        "Average Pivot Temperature",
        "Median Pivot Temperature",
        "Lower Pivot Range (25%)",
        "Upper Pivot Range (75%)"
    ],
    "Value": [
        pivot_candidates["temp_mean"].mean(),
        pivot_candidates["temp_mean"].median(),
        pivot_candidates["temp_mean"].quantile(0.25),
        pivot_candidates["temp_mean"].quantile(0.75)
    ]
})

pivot_summary.round(2)
```

The pivot summary table confirms that the temperature-driven booking shift happens in the mid-teen range. The average pivot temperature is 16.42°C, and the median is 16.05°C, which shows that the center of the transition is very close to 16°C. The middle 50% of pivot observations falls between 15.42°C and 17.90°C, so this range represents the most likely zone where customer preference changes from Heat-dominant to Ice-dominant demand.
