# Mapping the Barbering vs. Cosmetology Labor Landscape

```python
import pandas as pd
import numpy as np
```

## Load the dataset

```python
df = pd.read_csv("oews-2009-2025_original.csv")
df.head()
```

## Check the shape and columns

```python
print(df.shape)
print(df.columns.tolist())
```

## Clean the column names

```python
df.columns = (
    df.columns
    .str.strip()
    .str.replace(r"\s+", " ", regex=True)
)

for col in df.columns:
    print(repr(col))
```

## Check whether the two required SOC codes exist

In this dataset, the occupation codes appear without hyphens:
- 395011 = Barbers
- 395012 = Hairdressers, Hairstylists, and Cosmetologists

```python
target_codes = ["395011", "395012"]

filtered = df[df["Standard Occupational Classification"].astype(str).isin(target_codes)].copy()

print(filtered.shape)

filtered[[
    "Area Type",
    "Area Name",
    "Year",
    "Quarter",
    "Industry Name",
    "Standard Occupational Classification",
    "Occupational Title",
    "Number of Employed",
    "50th Percentile (Median) Wage"
]].head(20)
```

## Check the occupation counts

```python
print(filtered["Occupational Title"].value_counts())
```

## Check the available geography and time coverage

```python
print("Unique areas:", filtered["Area Name"].nunique())
print("Years:", sorted(filtered["Year"].dropna().unique())[:10], "...", sorted(filtered["Year"].dropna().unique())[-10:])
print("Area types:", filtered["Area Type"].value_counts())
```

# Step 2: Clean the OEWS dataset and create the focused subset

In this step, we keep the fields needed for the project, clean the numeric columns, and create a focused subset for the two required occupations.

## Keep only the columns needed for the project

```python
columns_needed = [
    "Area Type",
    "Area Name",
    "Year",
    "Quarter",
    "Industry Name",
    "Standard Occupational Classification",
    "Occupational Title",
    "Number of Employed",
    "Mean Wage",
    "50th Percentile (Median) Wage"
]

oews = df[columns_needed].copy()
oews.head()
```

## Check missing values in the selected fields

```python
print(oews.isnull().sum())
```

## Remove rows missing the core analysis fields

```python
oews = oews.dropna(subset=[
    "Area Type",
    "Area Name",
    "Standard Occupational Classification",
    "Occupational Title",
    "Number of Employed",
    "50th Percentile (Median) Wage"
]).copy()

print(oews.shape)
```

## Convert employment and wage fields to numeric

```python
numeric_cols = [
    "Number of Employed",
    "Mean Wage",
    "50th Percentile (Median) Wage"
]

for col in numeric_cols:
    oews[col] = pd.to_numeric(oews[col], errors="coerce")

print(oews[numeric_cols].dtypes)
```

## Remove rows that became invalid after numeric conversion

```python
oews = oews.dropna(subset=["Number of Employed", "50th Percentile (Median) Wage"]).copy()
print(oews.shape)
```

## Create the focused subset for the two target occupations

```python
target_codes = ["395011", "395012"]

focused = oews[oews["Standard Occupational Classification"].astype(str).isin(target_codes)].copy()
print(focused.shape)

focused[[
    "Area Type",
    "Area Name",
    "Year",
    "Quarter",
    "Industry Name",
    "Standard Occupational Classification",
    "Occupational Title",
    "Number of Employed",
    "50th Percentile (Median) Wage"
]].head(20)
```

## Create a short occupation group field

```python
def map_occ_group(code):
    code = str(code)
    if code == "395011":
        return "Barbers"
    elif code == "395012":
        return "Cosmetologists"
    else:
        return "Other"

focused["Occupation_Group"] = focused["Standard Occupational Classification"].astype(str).apply(map_occ_group)

focused[[
    "Standard Occupational Classification",
    "Occupational Title",
    "Occupation_Group"
]].drop_duplicates()
```

## Preview the cleaned focused subset

```python
print(focused.shape)
focused.head()
```

## Check the counts by occupation and year

```python
print(focused["Occupation_Group"].value_counts())
print(focused["Year"].value_counts().sort_index().tail(10))
```

# Step 3: Create Tableau-ready geographic and comparison fields

In this step, we create cleaner geography fields and a few helper fields that will make the Tableau dashboard easier to build.

## Create an area level field

This helps separate state-level rows from metropolitan-area rows.

```python
def map_area_level(area_type):
    a = str(area_type).strip().lower()

    if "state" in a:
        return "State"
    elif "metro" in a or "metropolitan" in a:
        return "Metropolitan Area"
    else:
        return "Other"

focused["Area_Level"] = focused["Area Type"].apply(map_area_level)
focused[["Area Type", "Area_Level"]].drop_duplicates()
```

## Create state and metro name helper fields

These make Tableau filters cleaner.

```python
focused["State_Name"] = np.where(
    focused["Area_Level"] == "State",
    focused["Area Name"],
    np.nan
)

focused["Metro_Name"] = np.where(
    focused["Area_Level"] == "Metropolitan Area",
    focused["Area Name"],
    np.nan
)

focused[["Area Name", "Area_Level", "State_Name", "Metro_Name"]].head(15)
```


## Create a unified geography field

This provides one general name field for labels and tooltips.

```python
focused["Geography_Name"] = focused["Area Name"]
focused[["Area Name", "Geography_Name"]].head()
```

## Rename the core employment and wage fields for easier use in Tableau

```python
focused = focused.rename(columns={
    "Number of Employed": "Employment",
    "50th Percentile (Median) Wage": "Median_Wage",
    "Mean Wage": "Mean_Wage",
    "Standard Occupational Classification": "SOC_Code"
})

focused.head()
```

## Create a simple wage gap helper field

This can support quick comparisons between median and mean wage if needed.

```python
focused["Wage_Gap"] = focused["Mean_Wage"] - focused["Median_Wage"]
focused[["Mean_Wage", "Median_Wage", "Wage_Gap"]].head()
```

## Preview the Tableau-ready subset

```python
print(focused.shape)
focused[[
    "Area Type",
    "Area_Level",
    "Area Name",
    "State_Name",
    "Metro_Name",
    "Year",
    "Quarter",
    "SOC_Code",
    "Occupation_Group",
    "Employment",
    "Median_Wage",
    "Mean_Wage"
]].head(20)
```

## Save the final Tableau-ready dataset

```python
focused.to_csv("barber_cosmetology_oews_ready.csv", index=False)
print("Saved as barber_cosmetology_oews_ready.csv")
```
