# Urban Canopy Data Consolidation Pipeline

In this project, I build a simple data pipeline to clean, standardize, and merge two different tree-related data sources.

The goal is to combine the city’s public street tree dataset with a separate volunteer planting log into one clean master inventory.

## Business Problem

The Urban Canopy Alliance works with two separate datasets:

- the city’s official street tree data
- the nonprofit’s volunteer planting log

Right now, these files are not easy to combine because they use different column names, formats, and levels of consistency.

The main business problem is to create a notebook-based pipeline that can clean both files, align their schemas, and merge them into one final inventory that is easier to use and maintain.

## Project Objective

The objective of this project is to automate the cleaning and merging of two tree datasets.

To do this, I will:
- load the public street tree dataset
- create a small mock volunteer log
- standardize column names
- clean dates and coordinates
- normalize species names
- merge both datasets into one master inventory
- export the final cleaned file

## Step 3: Create a More Realistic Mock Volunteer Log

The second data source in this project is a small mock volunteer planting log.

Instead of making it look too perfect, I create it in a more realistic way. The column names are different from the public tree dataset, the planting dates use mixed formats, and one row includes a missing coordinate on purpose.

This makes the cleaning and schema alignment step more realistic.

```python
volunteer_log = pd.DataFrame({
    "Species": [
        "coast redwood",
        "red flowering gum",
        "hybrid strawberry tree",
        "southern magnolia",
        "japanese maple",
        "london plane"
    ],
    "Planting Day": [
        "2024/01/14",
        "Jan 19 2024",
        "02-03-2024",
        "2024-02-18",
        "March 6, 2024",
        "2024/03/22"
    ],
    "Latitude_Value": [
        37.7749,
        37.7754,
        37.7760,
        37.7737,
        None,
        37.7728
    ],
    "Longitude_Value": [
        -122.4194,
        -122.4187,
        -122.4178,
        -122.4204,
        -122.4211,
        -122.4225
    ],
    "Volunteer_Group": [
        "Sunset Youth Crew",
        "Mission Green Team",
        "Richmond Weekend Volunteers",
        "Schoolyard Greening Day",
        "Neighborhood Tree Circle",
        "Civic Center Planting Crew"
    ],
    "Log_Notes": [
        "Planted along sidewalk edge",
        "Volunteer lead asked for mulch follow-up",
        "Replacement tree for damaged site",
        "Planted near playground fence",
        "Location written down but lat missing",
        "Final tree from Saturday event"
    ]
})

volunteer_log
```

## Why This Mock File Is Useful

This mock file overlaps with the public tree dataset, but it does not use the same headers.

For example:
- `Species` will later need to match the public species field
- `Planting Day` will need to be converted into a standard date format
- `Latitude_Value` and `Longitude_Value` will need to be aligned with the city coordinate fields

This makes it a better test for the consolidation pipeline.

## Step 4: Standardize the Column Names

At this point, the two datasets do not use the same column names, so they cannot be merged cleanly yet.

In this step, I rename the important fields in both datasets so they follow one shared naming style. This makes the next cleaning steps easier and prepares both files for merging into one master inventory.

```python
tree_data_standard = tree_data.rename(columns={
    "qSpecies": "common_name",
    "PlantDate": "planting_date",
    "Latitude": "latitude",
    "Longitude": "longitude"
})

volunteer_log_standard = volunteer_log.rename(columns={
    "Species": "common_name",
    "Planting Day": "planting_date",
    "Latitude_Value": "latitude",
    "Longitude_Value": "longitude"
})
```

## Check the Standardized Columns

After renaming the fields, I preview the main columns in both datasets to confirm that they now follow the same structure.

```python
tree_data_standard[["common_name", "planting_date", "latitude", "longitude"]].head()
```

```python
volunteer_log_standard[["common_name", "planting_date", "latitude", "longitude"]].head()
```

## Step 5: Clean the Planting Date Column

The two datasets now use the same column names, but the planting dates are still stored in different text formats.

In this step, I convert the `planting_date` column in both datasets into a consistent datetime format. This is important because the final pipeline should keep the planting date in a clean and standard form.

```python
tree_data_standard["planting_date"] = pd.to_datetime(
    tree_data_standard["planting_date"], errors="coerce"
)

volunteer_log_standard["planting_date"] = pd.to_datetime(
    volunteer_log_standard["planting_date"], errors="coerce"
)
```

## Check the Cleaned Date Column

After converting the dates, I check both datasets again to confirm that the `planting_date` column is now stored as a datetime value.

```python
tree_data_standard["planting_date"].dtype, volunteer_log_standard["planting_date"].dtype
```

```python
tree_data_standard[["common_name", "planting_date"]].head()
```

```python
volunteer_log_standard[["common_name", "planting_date"]].head()
```

## Fix the Date Parsing

The planting date column was converted into datetime format, but some values became missing (`NaT`) because the original date formats were mixed.

To make the volunteer log more consistent, I clean the text values and try the conversion again using a more flexible approach.

```python
volunteer_log_standard["planting_date"] = volunteer_log_standard["planting_date"].astype(str).str.strip()

volunteer_log_standard["planting_date"] = pd.to_datetime(
    volunteer_log_standard["planting_date"],
    errors="coerce"
)

volunteer_log_standard[["common_name", "planting_date"]]
```

## Fix the Volunteer Log Dates More Carefully

Some of the volunteer log dates use mixed formats, so the earlier conversion did not parse all of them correctly.

To fix this, I go back to the original volunteer log values and convert them again using a mixed-format parser.

```python
# rebuild the planting_date column from the original volunteer log values
volunteer_log_standard["planting_date"] = pd.to_datetime(
    volunteer_log["Planting Day"],
    format="mixed",
    errors="coerce"
)

volunteer_log_standard[["common_name", "planting_date"]]
```

## Step 6: Validate the Coordinate Columns

The next step is to clean the coordinate fields.

The project requires the final merged dataset to have zero missing values in the latitude and longitude columns, so I first remove rows that are missing either coordinate in both datasets.

```python
tree_data_clean = tree_data_standard.dropna(subset=["latitude", "longitude"]).copy()
volunteer_log_clean = volunteer_log_standard.dropna(subset=["latitude", "longitude"]).copy()
```

## Check the Coordinate Cleanup

After removing rows with missing latitude or longitude, I check the coordinate columns again to confirm that the remaining rows are complete.

```python
tree_data_clean[["latitude", "longitude"]].isnull().sum()
```

```python
volunteer_log_clean[["latitude", "longitude"]].isnull().sum()
```

## Step 7: Normalize the Species Names

The two datasets still use different writing styles for tree species names.

To make the final merged inventory more consistent, I now convert all species names into Title Case. This helps reduce small formatting differences before the merge.

```python
tree_data_clean["common_name"] = tree_data_clean["common_name"].astype(str).str.title().str.strip()
volunteer_log_clean["common_name"] = volunteer_log_clean["common_name"].astype(str).str.title().str.strip()
```

## Check the Normalized Species Names

After applying the formatting change, I preview the species names in both datasets to confirm that they now follow a more consistent style.

```python
tree_data_clean["common_name"].head()
```

```python
volunteer_log_clean["common_name"].head()
```

## Step 8: Merge Both Datasets into One Master Inventory

Now that both datasets have aligned column names, cleaned dates, complete coordinates, and normalized species names, I can merge them into one master inventory.

The project asks for an **outer join**, so I use that approach here to keep records from both sources in the final dataset.

```python
master_inventory = pd.merge(
    tree_data_clean,
    volunteer_log_clean,
    on=["common_name", "planting_date", "latitude", "longitude"],
    how="outer"
)
```

## Check the Merged Dataset

After merging the two sources, I check the shape and preview the first few rows to confirm that the final inventory was created successfully.

After confirming that the final merged dataset meets the main cleaning requirements, I export it to a CSV file named `master_inventory_2024.csv`.
