# Step 2: Clean the dataset and create the core operational fields

In this step, we standardize the dataset, convert the date field, ensure the numeric columns are usable, and create the first set of operational measures for supplier, cost, and waste analysis.

## Convert the date field to datetime

```python
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")
print(df["Date"].dtype)
```

## Remove rows with missing values in the core fields

```python
core_fields = [
    "Date",
    "Item_Name",
    "Category",
    "Current_Stock",
    "Daily_Usage",
    "Lead_Time",
    "Price_per_Unit",
    "Supplier_Name",
    "Seasonal_Factor",
    "Waste_Percentage"
]

df = df.dropna(subset=core_fields).copy()
print(df.shape)
```

## Convert the main numeric fields to numeric type

```python
numeric_cols = [
    "Current_Stock",
    "Reorder_Level",
    "Daily_Usage",
    "Lead_Time",
    "Price_per_Unit",
    "Seasonal_Factor",
    "Waste_Percentage"
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

print(df[numeric_cols].dtypes)
```

## Remove rows that became invalid after numeric conversion

```python
df = df.dropna(subset=numeric_cols).copy()
print(df.shape)
```

## Standardize the main text fields

```python
text_cols = ["Item_Name", "Category", "Subcategory", "Unit", "Supplier_Name"]

for col in text_cols:
    if col in df.columns:
        df[col] = df[col].astype(str).str.strip()

df.head()
```

## Create a total daily cost field

This approximates the daily purchasing value of each item.

```python
df["Daily_Cost"] = (df["Daily_Usage"] * df["Price_per_Unit"]).round(2)
df[["Daily_Usage", "Price_per_Unit", "Daily_Cost"]].head()
```

## Create a waste cost field

This estimates the cost lost to waste for each item.

```python
df["Waste_Cost"] = (
    df["Daily_Usage"] * df["Price_per_Unit"] * (df["Waste_Percentage"] / 100)
).round(2)

df[["Daily_Cost", "Waste_Percentage", "Waste_Cost"]].head()
```

## Create a stock pressure field

This helps identify items closer to stock risk.

```python
df["Stock_Pressure"] = (df["Reorder_Level"] - df["Current_Stock"]).round(2)
df[["Current_Stock", "Reorder_Level", "Stock_Pressure"]].head()
```

## Create a supplier performance proxy

Lower lead time suggests better supplier responsiveness.

```python
def supplier_performance_proxy(lead_time):
    if lead_time <= 2:
        return "High Performance"
    elif lead_time <= 5:
        return "Moderate Performance"
    else:
        return "Low Performance"

df["Supplier_Performance"] = df["Lead_Time"].apply(supplier_performance_proxy)
df[["Lead_Time", "Supplier_Performance"]].drop_duplicates().sort_values("Lead_Time")
```

# Step 3: Create seasonality, location, and dashboard helper fields

In this step, we derive month and season fields from the date, create a simulated location field for the four bistro branches, and add helper fields that will support the Tableau dashboard.

## Create month and weekday fields from the date

```python
df["Month"] = df["Date"].dt.month_name()
df["Weekday"] = df["Date"].dt.day_name()

df[["Date", "Month", "Weekday"]].head()
```

## Create a season field

This will support the seasonal volatility analysis in Tableau.

```python
def map_season(month_num):
    if month_num in [12, 1, 2]:
        return "Winter"
    elif month_num in [3, 4, 5]:
        return "Spring"
    elif month_num in [6, 7, 8]:
        return "Summer"
    else:
        return "Autumn"

df["Season"] = df["Date"].dt.month.apply(map_season)

df[["Date", "Season"]].tail(100)
```

## Create a simulated location field

The source dataset does not include restaurant branch location, so we create a clearly simulated four-location branch field for the dashboard.

```python
locations = [
    "Downtown",
    "Riverside",
    "Uptown",
    "Market District"
]

df["Location"] = [locations[i % len(locations)] for i in range(len(df))]
df["Location"].value_counts()
```

## Create a waste rate band

This makes it easier to visualize waste severity in Tableau.

```python
def waste_band(w):
    if w >= 15:
        return "High Waste"
    elif w >= 7:
        return "Moderate Waste"
    else:
        return "Low Waste"

df["Waste_Band"] = df["Waste_Percentage"].apply(waste_band)
df[["Waste_Percentage", "Waste_Band"]].head()
```

## Create a cost volatility proxy

This combines price and seasonal pressure into one simple volatility measure.

```python
df["Cost_Volatility_Score"] = (
    df["Price_per_Unit"] * df["Seasonal_Factor"]
).round(2)

df[["Price_per_Unit", "Seasonal_Factor", "Cost_Volatility_Score"]].head()
```

## Create a volume proxy field

This will support the cost vs volume toggle later in Tableau.

```python
df["Volume_Metric"] = df["Daily_Usage"].round(2)
df[["Daily_Usage", "Volume_Metric"]].head()
```

## Create a fulfillment proxy field

Since the source file does not include delivered vs ordered quantities, this gives a simple reliability-style proxy based on lead time and stock pressure.

```python
def fulfillment_proxy(row):
    if row["Lead_Time"] <= 2 and row["Stock_Pressure"] <= 0:
        return "Reliable"
    elif row["Lead_Time"] <= 5:
        return "At Risk"
    else:
        return "Unreliable"

df["Fulfillment_Proxy"] = df.apply(fulfillment_proxy, axis=1)
df[["Lead_Time", "Stock_Pressure", "Fulfillment_Proxy"]].head()
```
