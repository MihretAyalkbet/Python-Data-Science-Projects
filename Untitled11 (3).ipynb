```python
import pandas as pd
import random
from datetime import datetime, timedelta
```

```python
random.seed(42)
```

```python
n_rows = 150
```

## Define States, Regions, and UV Risk

These states are grouped into regions and assigned a UV risk level.

```python
states_regions_uv = {
    "Arizona": ("Southwest", "High"),
    "Nevada": ("Southwest", "High"),
    "Texas": ("South", "High"),
    "Florida": ("South", "High"),
    "California": ("West", "Medium"),
    "Colorado": ("West", "Medium"),
    "Georgia": ("South", "Medium"),
    "North Carolina": ("South", "Medium"),
    "Washington": ("West", "Low"),
    "Oregon": ("West", "Low"),
    "New York": ("Northeast", "Low"),
    "Massachusetts": ("Northeast", "Low")
}
```

## Define Categories

These are the values that will be used for vinyl type, finish, vehicle type, and issue type.

```python
vinyl_types = ["Monomeric", "Polymeric", "Cast"]
finishes = ["Gloss", "Matte", "Satin"]
vehicle_types = ["Sedan", "SUV", "Van", "Truck", "Fleet Van", "Trailer"]
issue_types_fade = ["Color Fade", "Discoloration"]
issue_types_edge = ["Edge Lifting", "Peeling", "Cracking"]

customer_names = [
    "Sun Fleet Logistics",
    "Desert Plumbing Co",
    "Metro Delivery Group",
    "Prime Wrap Transit",
    "Eco Route Services",
    "Canyon Fleet Media",
    "Urban Truck Ads",
    "Rapid Courier Systems",
    "BluePeak Installations",
    "Golden State Movers",
    "Skyline Transport",
    "East Coast Branding"
]
```

## Create Helper Functions

These functions help generate installation dates, issue scores, severity, and vinyl recommendations.

```python
def random_installation_date():
    today = datetime.today()
    start_date = today - timedelta(days=730)   # last 24 months
    random_days = random.randint(0, 730)
    return (start_date + timedelta(days=random_days)).date()

def random_claim_date(installation_date):
    max_claim_days = min(730, (datetime.today().date() - installation_date).days)
    if max_claim_days < 30:
        max_claim_days = 30
    claim_offset = random.randint(30, max_claim_days)
    return installation_date + timedelta(days=claim_offset)

def generate_issue_scores(uv_risk, vinyl_type, months_in_service):
    fade_score = 1
    edge_score = 1

    if uv_risk == "High":
        fade_score += 2
        edge_score += 2
    elif uv_risk == "Medium":
        fade_score += 1
        edge_score += 1

    if vinyl_type == "Monomeric":
        fade_score += 1
        edge_score += 1
    elif vinyl_type == "Polymeric":
        fade_score += 0.5
        edge_score += 0.5

    if months_in_service >= 18:
        fade_score += 1
        edge_score += 1
    elif months_in_service >= 10:
        fade_score += 0.5
        edge_score += 0.5

    fade_score = min(5, max(1, round(fade_score + random.uniform(-0.7, 0.7))))
    edge_score = min(5, max(1, round(edge_score + random.uniform(-0.7, 0.7))))

    return fade_score, edge_score

def get_severity(fade_score, edge_score):
    avg_score = (fade_score + edge_score) / 2
    if avg_score >= 4:
        return "High"
    elif avg_score >= 2.5:
        return "Medium"
    else:
        return "Low"

def get_recommended_grade(uv_risk):
    if uv_risk == "High":
        return "Cast Premium"
    elif uv_risk == "Medium":
        return "Premium Polymeric"
    else:
        return "Standard Polymeric"
```

## Generate the Synthetic Warranty Claims

Each row represents one customer warranty claim.

```python
rows = []

for i in range(1, n_rows + 1):
    state = random.choice(list(states_regions_uv.keys()))
    region, uv_risk = states_regions_uv[state]

    if uv_risk == "High":
        vinyl_type = random.choices(vinyl_types, weights=[0.4, 0.35, 0.25])[0]
    else:
        vinyl_type = random.choices(vinyl_types, weights=[0.3, 0.4, 0.3])[0]

    finish = random.choice(finishes)
    vehicle_type = random.choice(vehicle_types)

    installation_date = random_installation_date()
    claim_date = random_claim_date(installation_date)
    months_in_service = max(1, round((claim_date - installation_date).days / 30))

    fade_score, edge_score = generate_issue_scores(
        uv_risk, vinyl_type, months_in_service
    )

    severity = get_severity(fade_score, edge_score)
    recommended_grade = get_recommended_grade(uv_risk)

    if fade_score >= edge_score:
        issue_type = random.choice(issue_types_fade)
    else:
        issue_type = random.choice(issue_types_edge)

    rows.append({
        "Claim_ID": f"CLM{i:03d}",
        "Customer_Name": random.choice(customer_names),
        "State": state,
        "Region": region,
        "Installation_Date": installation_date,
        "Claim_Date": claim_date,
        "Months_in_Service": months_in_service,
        "Vinyl_Type": vinyl_type,
        "Finish": finish,
        "Vehicle_Type": vehicle_type,
        "Issue_Type": issue_type,
        "Reported_Color_Fade": fade_score,
        "Edge_Lifting": edge_score,
        "Severity": severity,
        "UV_Risk_Level": uv_risk,
        "Recommended_Vinyl_Grade": recommended_grade
    })
```

```python
df = pd.DataFrame(rows)
df.shape
```
