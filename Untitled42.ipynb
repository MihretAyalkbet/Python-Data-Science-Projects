# Predictive Sales Modeling for Whisk & Wander Pop-Up Festivals

In this project, I build a regression model to predict the quantity of dessert flights sold at outdoor festival pop-up events.

The goal is to understand how sales are affected by environmental conditions and event-related factors so the business can reduce waste and make better inventory decisions.

## Business Problem

Whisk & Wander operates dessert pop-up booths at outdoor botanical garden festivals.

Because the business works with delicate and perishable ingredients, it is important to predict sales as accurately as possible. If the expected sales are too high, the business may overprepare and waste ingredients. If the expected sales are too low, the business may miss sales opportunities.

The purpose of this project is to build a model that connects sales volume with weather and event-related conditions such as temperature, humidity, and festival duration.

The objective of this project is to:

## Step 3: Convert the Date Column and Generate Weather/Event Variables

The dataset now contains the main sales columns, but it does not yet include the environmental and event-related variables needed for the project scenario.

In this step, I first convert the `datetime` column into a proper datetime format. Then I create a few realistic additional features such as month, day number, temperature, humidity, festival duration, weekend activity, and special event days.

These generated variables will help simulate the outdoor pop-up conditions described in the business problem.

```python
festival_sales["datetime"] = pd.to_datetime(festival_sales["datetime"])

festival_sales["month"] = festival_sales["datetime"].dt.month
festival_sales["day_num"] = festival_sales["datetime"].dt.dayofweek
festival_sales["weekend_event"] = festival_sales["day_num"].isin([5, 6]).astype(int)

festival_sales[["datetime", "month", "day_num", "weekend_event"]].head()
```

## Generate Simulated Weather and Event Features

Since the original bakery sales dataset does not include weather fields, I create realistic external variables to match the project scenario.

These include:
- temperature
- humidity
- festival duration
- special event day indicator

They are generated in a simple but realistic way so the later regression model can learn from both sales history and outdoor event conditions.

```python
np.random.seed(42)

festival_sales["temperature"] = (
    18
    + 10 * np.sin(2 * np.pi * festival_sales["month"] / 12)
    + np.random.normal(0, 2, len(festival_sales))
).round(1)

festival_sales["humidity"] = (
    65
    + 15 * np.cos(2 * np.pi * festival_sales["month"] / 12)
    + np.random.normal(0, 5, len(festival_sales))
).round(1)

festival_sales["special_event_day"] = np.random.choice(
    [0, 1], size=len(festival_sales), p=[0.9, 0.1]
)

festival_sales["festival_duration_hours"] = np.where(
    festival_sales["weekend_event"] == 1,
    np.random.randint(4, 9, len(festival_sales)),
    np.random.randint(2, 6, len(festival_sales))
)

festival_sales[[
    "datetime",
    "temperature",
    "humidity",
    "festival_duration_hours",
    "weekend_event",
    "special_event_day"
]].head()
```

At this point, the dataset now contains both the original sales information and the generated weather/event variables that will be used later for the regression model.

## Step 7: Prepare the Data for Regression Modeling

After exploring the relationships in the data, the next step is to prepare the dataset for the regression model.

In this step, I select the main predictor variables and the target variable. The target is total sales, while the predictors include weather and event-related features such as temperature, humidity, festival duration, weekend activity, and special event days.

```python
features = festival_sales[[
    "temperature",
    "humidity",
    "festival_duration_hours",
    "weekend_event",
    "special_event_day"
]]

target = festival_sales["total"]

features.head()
```

## Split the Data into Training and Testing Sets

To evaluate the regression model fairly, I split the data into a training set and a testing set.

The model will learn patterns from the training data and then be evaluated on the test data.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    features, target, test_size=0.2, random_state=42
)

X_train.shape, X_test.shape
```

At this point, the data is ready for regression modeling.

The next step will be to train one or more regression models and compare their performance.

## Step 8: Train Regression Models

Now I train regression models to predict total sales using the weather and event-related features.

I begin with a simple **Linear Regression** model as a baseline, and then I also train a **Random Forest Regressor** to see whether a more flexible model performs better.

```python
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
```

### Model 1: Linear Regression

I first train a Linear Regression model to create a simple baseline for predicting sales.

```python
linear_model = LinearRegression()
linear_model.fit(X_train, y_train)

linear_predictions = linear_model.predict(X_test)
```

### Model 2: Random Forest Regressor

Next, I train a Random Forest Regressor, which can capture more complex patterns in the data than a simple linear model.

```python
forest_model = RandomForestRegressor(random_state=42, n_estimators=100)
forest_model.fit(X_train, y_train)

forest_predictions = forest_model.predict(X_test)
```

## Evaluate Model Performance

To compare the models, I use these regression metrics:

- **MAE (Mean Absolute Error)**: average size of prediction error
- **RMSE (Root Mean Squared Error)**: gives more weight to larger errors
- **R² Score**: shows how much of the variation in sales is explained by the model

```python
linear_mae = mean_absolute_error(y_test, linear_predictions)
linear_rmse = np.sqrt(mean_squared_error(y_test, linear_predictions))
linear_r2 = r2_score(y_test, linear_predictions)

forest_mae = mean_absolute_error(y_test, forest_predictions)
forest_rmse = np.sqrt(mean_squared_error(y_test, forest_predictions))
forest_r2 = r2_score(y_test, forest_predictions)

print("Linear Regression Performance")
print("MAE:", round(linear_mae, 2))
print("RMSE:", round(linear_rmse, 2))
print("R²:", round(linear_r2, 3))

print("\nRandom Forest Performance")
print("MAE:", round(forest_mae, 2))
print("RMSE:", round(forest_rmse, 2))
print("R²:", round(forest_r2, 3))
```

The results show that both models perform poorly, with negative R² scores indicating that they are not effectively capturing the relationship between the features and total sales. Linear Regression performs slightly better than Random Forest based on lower error values and a higher (though still negative) R² score. This suggests that the current features are not strong predictors of sales, and additional or more meaningful variables may be needed to improve model performance.

The next step is to compare the two models more clearly and identify which one gives the stronger sales prediction performance.

## Step 9: Compare the Models More Clearly

After evaluating both models, I compare them directly to see which one performed better.

This makes it easier to summarize the model results before moving to the final sales interpretation.

```python
model_comparison = pd.DataFrame({
    "Model": ["Linear Regression", "Random Forest Regressor"],
    "MAE": [linear_mae, forest_mae],
    "RMSE": [linear_rmse, forest_rmse],
    "R2": [linear_r2, forest_r2]
})

model_comparison
```

## Interpretation of the Model Results

The comparison table helps show which model performed better on the test data.

Even if the predictive performance is weak, this step is still useful because it tells me whether the current set of features is strong enough to explain sales or whether additional information may be needed.

## Improve the Feature Set

The first regression models performed poorly, which suggests that the generated weather and event variables alone were not enough to explain sales.

To improve the model, I now add stronger predictors from the original dataset, especially time-related information and location. These variables may capture more of the real sales pattern.

```python
festival_sales["place"] = festival_sales["place"].astype(str)
festival_sales_encoded = pd.get_dummies(festival_sales, columns=["place"], drop_first=True)

improved_features = festival_sales_encoded[[
    "temperature",
    "humidity",
    "festival_duration_hours",
    "weekend_event",
    "special_event_day",
    "month",
    "day_num"
] + [col for col in festival_sales_encoded.columns if col.startswith("place_")]]

target = festival_sales_encoded["total"]

improved_features.head()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    improved_features, target, test_size=0.2, random_state=42
)

X_train.shape, X_test.shape
```

## Retrain the Models with the Improved Feature Set

Now I train the regression models again using the improved feature set.

This version includes both the generated weather/event variables and stronger predictors from the original dataset, such as time patterns and place.

```python
linear_model = LinearRegression()
linear_model.fit(X_train, y_train)

linear_predictions = linear_model.predict(X_test)
```

```python
forest_model = RandomForestRegressor(random_state=42, n_estimators=100)
forest_model.fit(X_train, y_train)

forest_predictions = forest_model.predict(X_test)
```

## Evaluate the Improved Models

After retraining the models, I calculate the same evaluation metrics again to check whether the stronger feature set improves performance.

```python
linear_mae = mean_absolute_error(y_test, linear_predictions)
linear_rmse = np.sqrt(mean_squared_error(y_test, linear_predictions))
linear_r2 = r2_score(y_test, linear_predictions)

forest_mae = mean_absolute_error(y_test, forest_predictions)
forest_rmse = np.sqrt(mean_squared_error(y_test, forest_predictions))
forest_r2 = r2_score(y_test, forest_predictions)

print("Improved Linear Regression Performance")
print("MAE:", round(linear_mae, 2))
print("RMSE:", round(linear_rmse, 2))
print("R²:", round(linear_r2, 3))

print("\nImproved Random Forest Performance")
print("MAE:", round(forest_mae, 2))
print("RMSE:", round(forest_rmse, 2))
print("R²:", round(forest_r2, 3))
```

The improved model results show a small gain for Linear Regression, with the R² score moving slightly above zero. This suggests that the added time and place features helped explain sales a little better than before. However, the overall predictive performance is still weak, and the Random Forest model continues to perform worse than Linear Regression. This means the available features are still not strong enough to predict sales accurately, even after the improvement.

## Step 11: Compare the Improved Models

After retraining the models with a stronger feature set, I compare them again to see which one performs better.

This helps summarize the updated modeling results before moving to the final business interpretation.

```python
improved_model_comparison = pd.DataFrame({
    "Model": ["Improved Linear Regression", "Improved Random Forest Regressor"],
    "MAE": [linear_mae, forest_mae],
    "RMSE": [linear_rmse, forest_rmse],
    "R2": [linear_r2, forest_r2]
})

improved_model_comparison
```

## Interpretation of the Improved Model Results

The comparison table shows that the improved Linear Regression model performed slightly better than the Random Forest model.

Even though the improvement is small, this result suggests that the added time and location features helped the model capture a little more of the sales pattern than before.

## Step 12: Identify the Optimal Sales Window

The final business goal is to understand under which conditions sales are strongest.

To do that, I group the data into simple temperature and humidity ranges and then compare the average total sales across those groups. This helps turn the model results into a practical business insight.

```python
festival_sales["temp_group"] = pd.cut(
    festival_sales["temperature"],
    bins=[0, 10, 20, 30, 40],
    labels=["Cold", "Mild", "Warm", "Hot"]
)

festival_sales["humidity_group"] = pd.cut(
    festival_sales["humidity"],
    bins=[0, 50, 65, 80, 100],
    labels=["Low", "Moderate", "High", "Very High"]
)

festival_sales[["temperature", "temp_group", "humidity", "humidity_group"]].head()
```

## Average Sales by Temperature Range

This chart shows how average sales change across broad temperature conditions.

```python
temp_summary = festival_sales.groupby("temp_group", observed=False)["total"].mean().reset_index()

plt.figure(figsize=(8, 5))
sns.barplot(data=temp_summary, x="temp_group", y="total", color="seagreen")
plt.title("Average Sales by Temperature Range", fontsize=14)
plt.xlabel("Temperature Range", fontsize=12)
plt.ylabel("Average Total Sales", fontsize=12)
plt.show()
```

## Average Sales by Humidity Range

This chart shows how average sales vary across different humidity levels.

```python
humidity_summary = festival_sales.groupby("humidity_group", observed=False)["total"].mean().reset_index()

plt.figure(figsize=(8, 5))
sns.barplot(data=humidity_summary, x="humidity_group", y="total", color="steelblue")
plt.title("Average Sales by Humidity Range", fontsize=14)
plt.xlabel("Humidity Range", fontsize=12)
plt.ylabel("Average Total Sales", fontsize=12)
plt.show()
```

## Final Business Interpretation

The model results showed that the available features explain only a small part of the variation in total sales, even after improving the feature set.

However, the grouped sales summaries still provide practical business insight. By comparing average sales across temperature and humidity ranges, I can identify conditions that appear more favorable for stronger sales.

This helps the business think about its “optimal sales window” even if the regression model itself is not highly predictive.
